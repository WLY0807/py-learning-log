```python
# 输出格式美化
'''
Python两种输出值的方式: 表达式语句和 print() 函数。
第三种方式是使用文件对象的 write() 方法，标准输出文件可以用 sys.stdout 引用。
如果你希望输出的形式更加多样，可以使用 str.format() 函数来格式化输出值。
如果你希望将输出的值转成字符串，可以使用 repr() 或 str() 函数来实现。
str()： 函数返回一个用户易读的表达形式。面向人，输出简洁友好、可读性高，用于展示给用户看；
repr()： 产生一个解释器易读的表达形式。向Python 解释器 / 开发者，输出精准、完整，理论上复制输出字符串可以重新创建出原对象。
'''
s = "hello\nworld"

# str()：去掉外层引号，只展示可读文本，换行直接换行显示
print(str(s))
# 输出：
# hello
# world

# repr()：保留字符串完整字面形式，带引号、转义字符原样写出
print(repr(s))
# 输出：'hello\nworld'

# 两种方式输出一个平方与立方的表:
for x in range(1,6):
    print(repr(x).rjust(2),repr(x*x).rjust(3),end = '  ')
    print(repr(x**3).rjust(4))
'''
输出：
1   1     1
2   4     8
3   9    27
4  16    64
5  25   125

repr(x)
把数字转为带完整字面信息的字符串：
x=10 → repr(10) → 字符串 '10'

.rjust(width) 右对齐填充  ； .ljust(width) 左对齐填充
repr(x).rjust(2)：总宽度 2，数字右对齐，左边补空格
x=1 → ' 1'；x=10 → '10'

repr(x*x).rjust(3)：总宽度 3
平方 1 → ' 1'；平方 100 → '100'

'''

for x in range(1,6):
    print('{0:2d} {1:3d} {2:4d}'.format(x,x*x,x**3))
'''
输出：
1   1     1
2   4     8
3   9    27
4  16    64
5  25   125

格式模板语法 {序号:宽度d}
模板拆分 '{0:2d} {1:3d} {2:4d}'

{0:2d}
0：取 format 第 0 个参数 x
d：按十进制整数格式化
2：总宽度 2，数字默认右对齐，不足左边补空格

{1:3d}：取 format 里的第 1 个参数 x*x，宽度 3

{2:4d}：取 format 里的第 2 个参数 x*x*x，宽度 4
format 格式化字符串，{序号:宽度d} 统一控制数字排版，简洁规范；
'''

# 方法 zfill(), 它会在数字的左边填充 0，
# str.zfill(width)：在字符串左侧补 0，把字符串补齐到指定长度，返回新字符串，不修改原字符串。
s ='lj'
print(s.zfill(6))   # 0000lj
b = '-3.14'
print(b.zfill(6))   # -03.14,加填充的 0 一共六个字符
a = '12345678'
print(a.zfill(6))   # 12345678,若字符串本身长度超过 width 长度，则不做填充
```

```python
# str.format()
print('最有价值 {}： "{}!"'.format('球员', 'lj'))
# 输出：最有价值 球员： "lj!"
'''
大括号及其里面的字符 (称作格式化字段) 将会被 format() 中的参数替换。
在大括号中的数字用于指向传入对象在 format() 中的位置
'''
print('{0} 和 {1}'.format('MVP','FMVP'))   # MVP 和 FMVP
print('{1} 和 {0}'.format('MVP','FMVP'))   # FMVP 和 MVP

# 如果在 .format() 中使用了关键字参数, 那么它们的值会指向使用该名字的参数。
print('{honor} 球员：{name}'.format(honor = '最有价值',name = 'lj'))
# 输出：最有价值 球员：lj

# 位置及关键字参数可以任意的结合:
print('车站名 {0} ，{1}和 {other}'.format('north station','east station',other = 'zzzz'))
# 输出：车站名 north station ，east station和 zzzz
#担忧关键字的参数位置不能和没有参数的位置换位，此例中，other = 'zzzz' 与另外两个参数的位置不能换，否则会报错：
# SyntaxError: positional argument follows keyword argument

# : 和格式标识符可以跟着字段名。
# 在 : 后传入一个 .整数 , 可以保证该域至少有这么多的宽度,整数后的 字母 表示类型
import math
print('字符常量 PI 的值近似为：{0:.4f}'.format(math.pi))   # 字符常量 PI 的值近似为：3.1416

# 如果你有一个很长的格式化字符串, 而你不想将它们分开, 那么在格式化时通过变量名而非位置会是很好的事情。
# 最简单的就是传入一个字典，通过在 .format() 里的变量前使用 ** 来实现
table = {'Google':111,'Baidu':2222,'Edge':3333}
print('Baidu:{Baidu:d};Google:{Google:d};Edge:{Edge:d}'.format(**table))
# 输出：Baidu:2222;Google:111;Edge:3333
```

```python
# 读和写文件
'''
在 Python 中，文件操作通过 open() 函数完成。该函数会返回一个文件对象（file object），用于后续的读取或写入操作。
open() 的基本语法如下：
open(filename, mode)

filename：要访问的文件路径（字符串）。
mode：文件打开模式，用于指定读写方式（如只读、写入、追加等）。该参数是可选的，默认值为 r（只读）。

常见文件打开模式如下（建议重点掌握 r / w / a）：
模式	描述
r	    只读方式打开文件（默认模式），文件必须存在，文件指针位于开头。
rb	    以二进制格式只读打开文件（如图片、视频等），文件指针位于开头。
r+	    以读写方式打开文件，文件必须存在，文件指针位于开头。
rb+	    以二进制格式读写打开文件，文件指针位于开头。
w	    只写方式打开文件。如果文件存在会被清空；不存在则创建新文件。
wb	    以二进制格式只写打开文件。如果文件存在会被清空；不存在则创建新文件。
w+	    读写方式打开文件。如果文件存在会被清空；不存在则创建新文件。
wb+	    以二进制格式读写打开文件。如果文件存在会被清空；不存在则创建新文件。
a	    追加方式打开文件。如果文件存在，写入内容会追加到末尾；不存在则创建新文件。
ab	    以二进制格式追加写入文件，写入内容会追加到末尾；不存在则创建新文件。
a+	    读写方式打开文件，文件指针默认在末尾（追加模式）；不存在则创建新文件。
ab+	    以二进制格式读写打开文件，文件指针默认在末尾；不存在则创建新文件。
说明：在模式后添加 b 表示以二进制方式操作文件，常用于处理图片、音频等非文本文件。

下很好的总结了这几种模式：
open 打开文件
    ↓
操作类型
├─读取 → r 只读 → 指针:开头
├─写入 → 是否覆盖原内容
│        ├─是 → w 覆盖写
│        └─否 → a 追加写 → 指针:末尾
└─读写 → 是否覆盖原内容
         ├─是 → w+ 覆盖读写
         └─否 → 指针位置
                ├─开头 → r+ 读写
                └─末尾 → a+ 追加读写
'''
f = open("/mnt/c/Users/18371/Desktop/你好.txt","w")
f.write("python 是一个很好用的语言。\n对，的确非常好用!!\n")

# 关闭打开的文件
f.close()
'''
原本的文件路径是："C:/Users/18371/Desktop/你好.txt",(由于python 的语法中的正斜杠后有字符代表特殊含义，不能编译，所以其实原路径的所有斜杠和现在的是反着的)
在这里的文件路径是："/mnt/c/Users/18371/Desktop/你好.txt"
因为这是在 WSL2 环境下使用的，所以/mnt/c/ 是 WSL2 系统挂载机制，和 Python 语法无关；
Python 只是 “跟随当前操作系统的路径规则” 去找文件；
换环境就要换路径写法：Windows 原生用 C:/，WSL 内统一用 /mnt/c/。
'''
```

```python
# .tell()
'''
文件对象.tell()
用于返回文件当前的读/写位置（即文件指针的位置）。
文件指针表示从文件开头开始的字节数偏移量。f.tell() 返回一个整数，表示文件指针的当前位置。
'''

f = open("/mnt/c/Users/18371/Desktop/你好.txt","w")

print(f.tell())  # 0  刚打开，指针在0位置
f.write("abc")
print(f.tell())  # 3  写入3个英文字符，偏移量=3

f.close()
```

```python
# .seek()
'''
如果要改变文件指针当前的位置, 可以使用 f.seek(offset, from_what) 函数。
f.seek(offset, whence) 用于移动文件指针到指定位置。
offset 表示相对于 whence 参数的偏移量，from_what 的值, 如果是 0 表示开头, 如果是 1 表示当前位置, 2 表示文件的结尾
例如：
seek(x,0) ： 从起始位置即文件首行首字符开始移动 x 个字符
seek(x,1) ： 表示从当前位置往后移动x个字符
seek(-x,2)：表示从文件的结尾往前移动x个字符
from_what 值为默认为0，即文件开头。
'''

f = open("/mnt/c/Users/18371/Desktop/你好.txt","rb+")

f.write(b'0123456789abcdef')  # 16

f.seek(5)  # 5  ,移动到文件的第六个字节，为 5

f.read(1)  # b'5'

f.seek(-3,2)   # 13 ,移动到文件的倒数第三字节

f.read(1)    # b'd'

'''
字符串前加 b'xxx' 的含义
b 代表 bytes（字节串），区别于普通 str 字符串

为什么你这里必须写 b''
你打开文件用的模式是 rb+：二进制读写模式
二进制模式下，write() 只能接收 bytes 类型，不能传普通 str；
如果直接写 f.write("0123") 会直接报错，必须转成字节串

补充区分两种打开模式
不带 b：r/w/a/r+ 文本模式
读写用 str 字符串，自动处理编码换行；
不能写 b''，直接传文本即可。

带 b：rb/wb/ab/rb+ 二进制模式
读写必须用 bytes 字节串，必须加 b'' 或 .encode()；
.tell()、.seek() 全部按字节计算
'''
```

```python
# pickle 模块
'''
python的pickle模块实现了基本的数据序列和反序列化。
通过pickle模块的序列化操作我们能够将程序中运行的对象信息保存到文件中去，永久存储。
通过pickle模块的反序列化操作，我们能够从文件中创建上一次程序保存的对象。

基本接口：
pickle.dump(obj, file, [,protocol])
1. 作用
将 Python 对象 obj 序列化，直接写入打开的二进制文件对象 file，把对象永久保存到本地文件。
序列化：把列表、字典、元组、自定义类实例等复杂 Python 对象，转成二进制字节存入文件，后续可用 pickle.load() 还原出原对象。
2. 参数说明
参数 1：obj
任意 Python 对象：list /dict/tuple / 自定义类实例 / 函数等
参数 2：file
必须是以二进制写入模式打开的文件对象（wb / ab / rb+）
不能用文本模式 w/r，否则报错。
参数 3：protocol（可选，整数）
序列化协议版本，数字越大效率越高、兼容性越差：
0：纯文本，兼容所有 Python
4：Python3.4+ 常用
5：Python3.8+，支持超大数组
省略时自动使用当前 Python 默认协议。

有了 pickle 这个对象, 就能对 file 以读取的形式打开:
x = pickle.load(file)
# 从 file 中读取一个字符串，并将它重构为原来的python对象。
file: 类文件对象，有read()和readline()接口。
'''

# 实例 1  创建 pickle 文件

# 导入序列化模块，用于保存/还原Python对象
import pickle

# 使用 pickle 模块将数据对象保存到文件

# 定义字典对象，包含数字、浮点数、复数、字符串、空值等多种类型
data1 = {'a':[1,2.0,3,4+6j],
        'b': ('string',u'Unicode string'),
        'c':None
        }

# 定义自引用列表（列表内部包含自己，循环嵌套）
selfref_list = [1,2,3]
selfref_list.append(selfref_list)
'''
这段自循环列表 selfref_list 是用来演示 pickle 独有的强大特性;
pickle 处理自循环对象的好处（这段代码的教学目的）
好处 1：无损保存循环 / 自引用结构，完整还原
好处 2：演示 pickle 能处理复杂特殊对象
日常开发很少主动写自引用，但这个例子能体现 pickle 不可替代的优势。
'''

# 以二进制写入模式打开文件，必须带 wb（pickle只支持二进制模式），data 为文件名
output = open('data.pkl','wb')
# 第一个对象：用默认协议0序列化字典写入文件
pickle.dump(data1,output)

# 第二个对象：protocol=-1 代表使用当前Python支持的最高版本协议，压缩更好、速度更快
pickle.dump(selfref_list,output,-1)

# 关闭文件释放资源
output.close()
```

```python
# 实例2  打开 pickle 文件

import pprint,pickle
# pickle：读取pkl文件、还原对象
# pprint：格式化打印，让字典/列表排版更清晰，比print好看

# 二进制只读模式打开之前生成的data.pkl文件
pkl_file = open('data.pkl','rb')

# 第一次load：读取第一个存入的对象（字典data1）
data1 = pickle.load(pkl_file)
pprint.pprint(data1)
# 格式化打印字典

# 第二次load：读取第二个存入的对象（自引用列表selfref_list）
data2 = pickle.load(pkl_file)
pprint.pprint(data2)
# 格式化打印列表

# 关闭文件，释放资源
pkl_file.close()

'''
输出：
{'a': [1, 2.0, 3, (4+6j)], 'b': ('string', 'Unicode string'), 'c': None}
[1, 2, 3, <Recursion on list with id=127859621805760>]’
'''
```
