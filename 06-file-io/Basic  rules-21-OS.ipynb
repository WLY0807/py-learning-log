```python
# OS 文件/目录方法
'''
os 模块是 Python 标准库中的一个重要模块，它提供了与操作系统交互的功能。
通过 os 模块，你可以执行文件操作、目录操作、环境变量管理、进程管理等任务。
os 模块是跨平台的，这意味着你可以在不同的操作系统（如 Windows、Linux、macOS）上使用相同的代码。
'''
```

```python
# OS 模块的常用函数

# 1. 获取当前工作目录
# os.getcwd() 
# 函数用于获取当前工作目录的路径。当前工作目录是 Python 脚本执行时所在的目录。

import os

current_directory = os.getcwd()
print("当前工作目录:", current_directory)
# 当前工作目录: /home/wsl2/py-learning-log
```

```python
# 2. 改变当前工作目录
# os.chdir(path) 
# 函数用于改变当前工作目录。path 是你想要切换到的目录路径。

import os

os.chdir("/home/wsl2/py-learning-log")
print("新的工作目录:", os.getcwd())
# 新的工作目录: /home/wsl2/py-learning-log
# 这里我不想更改我的路径，所以 path 还是原来的路径
```

```python
# 3. 列出目录内容
# os.listdir(path) 
# 函数用于列出指定目录中的所有文件和子目录。如果不提供 path 参数，则默认列出当前工作目录的内容。

import os

files_and_dirs = os.listdir("/home/wsl2")
print("目录内容:", files_and_dirs)
'''
目录内容: ['.anaconda',  '.jupyter']
'''
```

```python
# 4. 创建目录
# os.mkdir(path) 
# 函数用于创建一个新的目录。如果目录已经存在，会抛出 FileExistsError 异常。

import os

os.mkdir("new_directory")
'''
可以看到左侧的参数列表，多了一个名为 new_directory 的文件夹
'''
```

```python
# 5. 删除目录
# os.rmdir(path) 
# 函数用于删除一个空目录。如果目录不为空，会抛出 OSError 异常。

import os

os.rmdir("new_directory")
# 可以看到左侧的参数列表刚建好的 new_directory 的空列表已将被删除了
```

```python
# 6. 删除文件
# os.remove(path) 
# 函数用于删除一个文件。如果文件不存在，会抛出 FileNotFoundError 异常。

import os

os.remove("/mnt/c/Users/18371/Desktop/test.txt")
#  在我桌面的一个名为 test 的文本文件已将被删除了
```

```python
# 7. 重命名文件或目录
# os.rename(src, dst) 
# 函数用于重命名文件或目录。src 是原始路径，dst 是新的路径。

import os

os.rename("/home/wsl2/py-learning-log/测试文本.txt", "/home/wsl2/cpp-learning-log/测试文本.txt")
'''
可以看到左侧的 一个名为 测试文本的 .text 文件已被移到 cpp-learning-log 文件夹里了
还可以只修改名字，不改地址：
old = "/home/wsl2/py-learning-log/测试文本.txt"
new = "/home/wsl2/py-learning-log/新名字.txt"
os.rename(old, new)
'''
```

```python
# 8. 获取环境变量
# os.getenv(key) 
# 函数用于获取指定环境变量的值。如果环境变量不存在，返回 None。
# key: 字符串，环境变量名称（大小写区分，Linux 严格区分，Windows 不严格）
import os

home_directory = os.getenv("HOME")
print("HOME 目录:", home_directory)
'''
输出：
目录: /home/wsl2

WSL/Linux 系统环境变量：HOME、PATH、USER、LANG
Windows 常用：USERNAME、USERPROFILE、SystemRoot
环境变量全部是字符串，数字读取后需要手动转 int/float；
os.getenv 不会修改系统环境，仅读取。
'''
```

```python
# 9. 执行系统命令
# os.system(command) 
# 函数用于在操作系统的 shell 中执行命令。命令执行后，返回命令的退出状态。

import os

os.system("pwd")
'''
输出：
/home/wsl2/py-learning-log

WSL / Linux 常用终端命令（配合 os.system () 使用
目录查看类
bash
运行
ls          # 列出当前目录文件
ls -l       # 详细信息（权限、大小、时间）
ls -a       # 显示隐藏文件
ls -lh      # 人性化显示文件大小(K/M/G)
pwd         # 查看当前所在完整路径

2. 切换 / 创建 / 删除文件夹
bash
运行
cd 目录名         # 进入文件夹
cd ..             # 返回上一级
mkdir test        # 创建单层文件夹
mkdir -p a/b/c    # 递归创建多层文件夹
rmdir test        # 删除空文件夹
rm -rf test       # 强制删除文件夹+内部所有文件（谨慎！）

3. 文件操作
bash
运行
touch a.txt           # 新建空文件
rm a.txt              # 删除文件
cp a.txt b.txt        # 复制文件
cp -r folder1 folder2 # 复制文件夹
mv old.txt new.txt    # 重命名文件
mv a.txt /home/xxx/   # 移动文件到目标目录

4. 文件查看、读写
bash
运行
cat a.txt          # 一次性打印全部内容
head a.txt         # 只看前10行
tail a.txt         # 只看末尾10行
tail -n 3 a.txt    # 查看最后3行

5. 文件搜索
bash
运行
find . -name "*.txt"   # 当前目录查找所有txt文件
grep "关键词" a.txt     # 在文件中搜索文字

6. 系统信息
bash
运行
whoami        # 当前用户名
df -h         # 磁盘占用
free -h       # 内存使用
clear         # 清空终端屏幕

7. 压缩解压
bash
运行
tar -zcvf 打包名.tar.gz 文件夹/文件  # 压缩
tar -zxvf 打包名.tar.gz              # 解压

二、Windows CMD 常用命令（Windows 系统 os.system 用）
cmd
dir                 # 列出当前目录文件（替代ls）
cd                  # 查看当前路径
cd 文件夹名         # 进入目录
md test             # 创建文件夹
rd test             # 删除空文件夹
rd /s /q test       # 强制删除文件夹
type a.txt          # 查看文件内容（替代cat）
del a.txt           # 删除文件
copy a.txt b.txt    # 复制
move old.txt new.txt # 重命名/移动
cls                 # 清屏
'''
```

```python
# 10.os.makedirs(path)
'''
递归创建多层级文件夹，路径里所有不存在的父目录会一次性全部创建。
os.makedirs(path, mode=0o777, exist_ok=False)
path：要创建的文件夹完整路径
mode：文件夹权限（Linux/WSL 生效，Windows 忽略）
exist_ok=False：默认，如果文件夹已存在，报错 FileExistsError；设为 True 存在也不报错
'''
import os

path = "/mnt/c/Users/18371/Desktop/a/b/c"
os.makedirs(path)
# 此时我的桌面上已经多了一个名为 a 的文件夹 里外一共嵌套了三层
```

```python
# 11.os.path.join(a,b)
'''
拼接路径片段，自动适配系统分隔符：
Linux/WSL：/
Windows：\

语法：
full_path = os.path.join(路径1, 路径2, 路径3...)
支持传入多个参数，依次拼接。
'''
import os

# 基础目录
desktop = "/mnt/c/Users/18371/Desktop"
txt_name = "你好.txt"

# 拼接完整路径
file_path = os.path.join(desktop, txt_name)
print("完整路径为：",file_path)      # 完整路径为： /mnt/c/Users/18371/Desktop/你好.txt



# 12.os.path.exists(path)
'''
判断指定路径是否存在，返回布尔值：
True：路径存在（不管是文件 / 文件夹都算存在）
False：路径不存在

语法
result = os.path.exists(路径字符串)
'''
if os.path.exists(file_path):
    fo = open(file_path, "r", encoding="utf-8")
    print(fo.read())
    fo.close()
'''
输出：
0123456789abcdef
我爱你
我不敢对你保证
表示此文件存在
'''


# 13.os.path.isfile(path)
'''
判断给定路径是否是普通文件，返回布尔值：
True：路径存在，且是文件（txt、pkl、png 等）
False：不存在 / 是文件夹 / 软链接损坏
'''
if os.path.isfile(file_path):
    fol = open(file_path, "r", encoding="utf-8")
    print(fol.read())
    fol.close()
'''
输出：
0123456789abcdef
我爱你
我不敢对你保证
表示这就是一个普通文件
'''

# 14.os.path.isdir(path)
'''
判断路径是否为文件夹（目录），返回布尔值：
True：路径存在，且是文件夹
False：路径不存在 / 是文件 / 损坏快捷方式
'''
if os.path.isdir(file_path):
    foll = open(file_path, "r", encoding="utf-8")
    print(foll.read())
    foll.close()
else:
    print("这不是一个常规文件夹")
# 这不是一个常规文件夹  ;表明这不是文件夹

'''
函数	                返回 True 条件
os.path.exists(path)	文件、文件夹只要存在就为 True
os.path.isfile(path)	存在，且是普通文件（txt/pkl 等）
os.path.isdir(path)	    存在，且是文件夹
'''
```
