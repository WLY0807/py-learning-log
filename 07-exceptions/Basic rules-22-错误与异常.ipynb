```python
# 异常处理
# try/except 语句
'''
异常捕捉可以使用 try/except 语句。
try:
    执行代码
except：
    发生异常时执行的代码
'''
# 以下例子中，让用户输入一个合法的整数，但是允许用户中断这个程序（使用 Control-C 或者操作系统提供的方法）。
# 用户中断的信息会引发一个 KeyboardInterrupt 异常。
while True:
    try:
        x = int(input("请输入一个数字："))
        break
    except ValueError:
        print("您输入的不是数字，请再次尝试输入！")
'''
try 语句按照如下方式工作；
首先，执行 try 子句（在关键字 try 和关键字 except 之间的语句）。
如果没有异常发生，忽略 except 子句，try 子句执行后结束。
如果在执行 try 子句的过程中发生了异常，那么 try 子句余下的部分将被忽略。如果异常的类型和 except 之后的名称相符，那么对应的 except 子句将被执行。
如果一个异常没有与任何的 except 匹配，那么这个异常将会传递给上层的 try 中。

一个 try 语句可能包含多个except子句，分别来处理不同的特定的异常。最多只有一个分支会被执行。
处理程序将只针对对应的 try 子句中的异常进行处理，而不是其他的 try 的处理程序中的异常。
# 一个except子句可以同时处理多个异常，这些异常将被放在一个括号里成为一个元组:
except (RuntimeError, TypeError, NameError):
'''
# 最后一个except子句可以忽略异常的名称，它将被当作通配符使用
# 可以使用这种方法打印一个错误信息，然后再次把异常抛出。

import sys

try:
    f = open('测试文本1.txt')
    s = f.readline()
    i = int(s.strip())    # .strip() ：移除字符串首尾空白字符（空格、\n、\t、\r）。
except OSError as err:
    print("OS error:{0}".format(error))
except ValueError:
    print("Could not convert data to an integer.")
except:
    print("Unexpected error:",sys.exc_info()[0])
    raise
    f.close()
# 输出：Could not convert data to an integer.
# 因为。text文件里的是 文字，不是数字
'''
 三层 except 捕获顺序（从上到下匹配）
1. except OSError as err
捕获和操作系统、文件 IO 相关异常：
FileNotFoundError 文件找不到
PermissionError 权限不足
FileNotFoundError 继承自 OSError，可以被这条捕获。
2. except ValueError
捕获字符串无法转为数字的异常。
例：文件第一行内容：hello → int("hello") 报错。
3. except: 【万能捕获】
没有写异常类型，捕获所有剩下未匹配的异常
sys.exc_info()[0]：获取当前异常的类型；这个函数有三个参数，对应 0 异常类型, 1 异常实例, 2 追踪对象
末尾 raise：重新抛出异常
作用：打印异常类型信息后，不让程序静默吞掉错误，程序依旧终止。
'''
```

```python
# try/except...else 语句
'''
try/except 语句还有一个可选的 else 子句，如果使用这个子句，那么必须放在所有的 except 子句之后。
else 子句将在 try 子句没有发生任何异常的时候执行。
try:
    执行代码
except:
     发生异常时执行的代码
else:
     没有异常时执行的代码
'''

# 以下实例在 try 语句中判断文件是否可以打开，如果打开文件时正常的没有发生异常则执行 else 部分的语句，读取文件内容：
import sys

for arg in sys.argv[1:]:
    try:
        with open(arg, 'r', encoding="utf-8") as f:
            lines = f.readlines()
    except OSError:
        print('cannot open', arg)
    else:
        print(arg, 'has', len(lines), 'lines')
'''
输出：
cannot open -f
/home/wsl2/.local/share/jupyter/runtime/kernel-d606ef02-88e2-4cf2-8c4f-a6835266fdb7.json has 13 lines
因为可能运行脚本传入的命令行参数大概类似：python xxx.py -f /home/wsl2/.local/share/jupyter/runtime/kernel-d606ef02-88e2-4cf2-8c4f-a6835266fdb7.json
所以循环依次遍历的第一轮 arg = "-f"  ，系统尝试打开一个名叫 -f 的文件，不存在 → 触发 OSError

第二轮 arg = "/home/wsl2/.local/kernel-xxx.json"
文件正常打开，读取成功，输出：
plaintext
/home/wsl2/.local/share/jupyter/runtime/kernel-d606ef02-88e2-4cf2-8c4f-a6835266

sys.argv 是命令行参数列表
sys.argv[0]：当前运行的脚本文件名
sys.argv[1:]：切片，取出脚本名之后所有传入的参数

for arg in sys.argv[1:]:
arg 依次代表每一个传入的文件路径。

encoding="utf-8
文本模式r下，不指定编码，Python 会使用操作系统默认编码
WSL/Linux 默认 UTF-8；打开带中文的文件极易乱码、报UnicodeDecodeError。
'''
```

```python
# try-finally 语句
'''
try-finally 语句无论是否发生异常都将执行最后的代码。
try:
    执行代码
except:
    发生异常时执行的代码
else:
    没有异常时执行的代码
finally：
    无论有无异常都会执行的代码
'''

try:
    with open("/mnt/c/Users/18371/Desktop/你好.txt") as f:
        read = f.read()
        print(read)
except AssertionError as error:
    print(error)
else:
    try:
        with open('测试文本1') as file:
            read_dates = file.read()
    except FileNotFoundError as fnf_error:
        print(fnf_error)
finally:
    print('无论如何，你都会看到这句话')
'''
输出：
0123456789abcdef
我爱你
我不敢对你保证

[Errno 2] No such file or directory: '测试文本1'
无论如何，你都会看到这句话
'''
```

```python
# raise --抛出异常
'''
Python 使用 raise 语句抛出一个指定的异常。

raise语法格式如下：
raise [Exception [, args [, traceback]]]

raise 唯一的一个参数指定了要被抛出的异常。它必须是一个异常的实例或者是异常的类（也就是 Exception 的子类）。
raise 不带任何参数，是 Python 专门设计的语法糖。
Python 解释器会自动记住：当前正在处理的异常对象，直接把这个缓存的原始异常重新向上抛出。
'''
'''
# 实例1
x = 10
if x > 5:
    raise Exception('x 不能大于5，此时 x 的值为：{}'.format(x))

输出：
Exception                                 Traceback (most recent call last)
Cell In[8], line 14
     12 x = 10
     13 if x > 5:
---> 14     raise Exception('x 不能大于5，此时 x 的值为：{}'.format(x))

Exception: x 不能大于5，此时 x 的值为：10
'''

# 实例2
# 如果只想知道这是否抛出了一个异常，并不想去处理它，那么一个简单的 raise 语句就可以再次把它抛出。
try:
    raise NameError('HiThere')    # 模拟一个异常。
except NameError:
    print('An exception flew by!')
    raise  # 如果不单独抛出这个 raise 的话，输出就会只有 ：An exception flew by!
'''
输出：
An exception flew by!
---------------------------------------------------------------------------
NameError                                 Traceback (most recent call last)
Cell In[14], line 2
      1 try:
----> 2     raise NameError('HiThere')    # 模拟一个异常。
      3 except NameError:
      4     print('An exception flew by!')
      
NameError: HiThere

代码执行流程推演
进入 try 代码块；
执行 raise NameError('HiThere')，主动抛出 NameError 异常；
异常匹配 except NameError，进入分支；
打印文本：An exception flew by!；
执行单独 raise，再次向外抛出刚刚捕获的原始 NameError 异常；
外层没有额外 try-except 捕获，程序终止，打印完整异常堆栈。
'''
```

```python
# 用户自定义异常
'''
可以通过创建一个新的异常类来拥有自己的异常。
异常类继承自 Exception 类，可以直接继承，或者间接继承
'''
class MyError(Exception):
    def __init__(self,value):
        self.value = value
    def __str__(self):    # 魔术方法（字符串转换方法）
        return repr(self.value)

try:
    raise MyError(2*2)
except MyError as e:
    print('My exception occurred,value:',e.value)

# 输出：My exception occurred,value: 4

raise MyError('error')
'''
输出：
MyError                                   Traceback (most recent call last)
Cell In[17], line 20
     15     print('My exception occurred,value:',e.value)
     16 
     17 输出：
     18 My exception occurred,value: 4
     19 
---> 20 raise MyError('error')
MyError: 'error'
在这个例子中，类 Exception 默认的 __init__() 被覆盖。

repr() 内置函数
作用：返回对象的 “官方字符串表示形式”。这里他让返回的异常类型 内容加上了 ''
repr(4) → '4'；如果是字符串 repr("abc") → "'abc'"。

在这个异常类里定义的魔法函数参数规则：
1.仅允许 self、不能追加参数的魔法函数：__str__、__repr__、__len__、__bool__、__iter__、__next__
2.固定接收一个外部参数 (other/key/item/name)：运算符、索引、属性相关魔法方法
3.参数自由定制：__init__、__new__、__call__
'''
```

```python
'''
当创建一个模块有可能抛出多种不同的异常时，一种通常的做法是：
为这个包建立一个基础异常类，然后基于这个基础类为不同的错误情况创建不同的子类
'''

class Error(Exception):
    '''Base class for exceptions in this moudle'''
       '''本模块中异常的基类。'''
    pass

class InputError(Error):
    '''Exception raised for errors in the input.

    Attributes:
       expression --input expression in which the error occured
       message --explanation of the error
    '''
    '''由于输入存在错误而引发了异常。
    属性说明：expression——出现错误的输入表达式；message——错误的原因说明。'''

    def __init__(self,expression,message):
        self.expression = expression
        self.message = message

class TransitionError(Error):
    '''Raised when an operation attempts asttate transition that's not
    allowed.

    Attributes:
        previous --state at beginning of transition
        next --attempted new state
        message --explanation of why the specific transition is not allowed
    '''
    '''当某个操作试图进行不被允许的状态转换时，就会触发此异常。
    相关属性包括：previous——状态转换开始时的状态；next——试图达到的新状态；message——说明为何该状态转换不被允许的原因。'''
    
def __init__(self,previous,next,message):
    self.previous = previous
    self.next = next
    self.message = message
'''
设计思想（工程标准写法）
先定义一个基础父异常 Error；
根据业务场景派生多个细分异常（InputError、TransitionError）；
每个子类 __init__ 按需定义参数，保存业务相关数据；
调用方可以：
精准捕获某一类异常（except InputError）
统一捕获全部模块内异常（except Error
大多数的异常的名字都以"Error"结尾，就跟标准的异常命名一样。
'''
```
