```python
class Solution:
    def isPalindrome(self, x: int) -> bool:
        x_max = (2**31) - 1
        x_min = -(2**31)
        res = 0
        original_x = x
        if (x > 0) and (x < x_max) :
            while x != 0:
                digit = x % 10
                x = x // 10
                res =(res * 10)  + digit 
                if res == original_x:
                    return True
        if original_x == 0:
            return True
        else:
            return False 

obj = Solution()
print(obj.isPalindrome(0))
```

```python
from typing import List
class Solution:
    def removeElement(self, nums: List[int], val: int) -> int:
        i = 0
        while i < len(nums) :
            if nums[i] == val:
                nums.pop(i)
            else :
                i += 1
        return len(nums)

nums = [1,1,1,3,3,2,5,8,8,4,6,3]
obj = Solution()
obj.removeElement(nums,3)  # 9
print(nums)
```
