```python
# 循环语句
'''
Python 中的循环语句有 for 和 while。
关键字 / 函数	说明	                            示例
for	            迭代循环，用于遍历序列或可迭代对象	    for i in list:
while	        条件循环，条件为 True 时持续执行	    while x > 0:
break	        立即终止当前循环	                    break
continue	    跳过本次循环剩余代码，进入下一次迭代	continue
else（循环）    	循环正常结束（未被 break）时执行	    for i in range(3): ... else: ...
pass	        循环中的占位语句（空操作）	        for i in range(5): pass
range()	        生成整数序列，常与 for 循环配合使用	range(0, 5)
enumerate()	    遍历时同时获取索引和值	            for i, v in enumerate(list):
'''
```

```python
# while 循环
'''
Python 中 while 语句的一般形式：
while 判断条件(condition)：
    执行语句(statements)……
同样需要注意冒号和缩进。另外，在 Python 中没有 do..while 循环。
'''
# 以下实例使用了 while 来计算 1 到 100 的总和：
n = 100

sum = 0
counter = 1
while counter <= n:
    sum += counter
    counter += 1
    
print("1到 %d 之和为：%d" % (n,sum))

# while 循环使用 else 语句
'''
如果 while 后面的条件语句为 false 时，则执行 else 的语句块:
while <expr>:
    <statement(s)>
else:
    <additional_statement(s)>
'''
count = 0
while count < 5:
    print (count ,"小于5")
    count += 1
else:
    print(count ,"大于或等于5")
```

```python
# for 语句
'''
Python for 循环可以遍历任何可迭代对象，如一个列表或者一个字符串:
for <variable> in <sequence>:    # 循环每一轮，自动把 sequence 里下一个元素赋值给 variable
    <statements>                 # 不需要用到变量时，约定variable处用下划线 _ 占位
else:
    <statements>
'''
sites = ["baidu","google","jingdong","taobao"]
for site in sites:
    print(site)

word = "hello"
for letter in word:
    print(letter,'  ',end = '')   # end = '' ，可让输出不换行

print()  # 空输出等于换行

#  1 到 5 的所有数字
for number in range(8,18):
    print(number,' ',end = '')
    
print()

# for...else 语句用于在循环结束后执行一段代码。
# 当循环执行完毕（即遍历完 iterable 中的所有元素）后，会执行 else 子句中的代码，
# 如果在循环过程中遇到了 break 语句，则会中断循环，此时不会执行 else 子句。
# break 语句可以跳出 for 和 while 的循环体。如果你从 for 或 while 循环中终止，任何对应的循环 else 块将不执行。
# continue 语句被用来告诉 Python 跳过当前循环块中的剩余语句，然后继续进行下一轮循环。

for x in range(8):
    print(x,' ',end = '')
else:
    print("Finished!")

sites = ["baidu","google","jingdong","taobao"]
for site in sites:
    if site == "jingdong":
        print("京东跳出循环！")
        break
    print("循环数据：" + site)
else:
    print("无循环数据！")
print("完成循环！")

print("------------------------------")

sites = ["baidu","google","jingdong","taobao"]
for site in sites:
    if site == "jingdong":
        print("京东跳出循环！")
        continue
    print("循环数据：" + site)
else:
    print("无循环数据！")
print("完成循环！")

# Python中 pass是空语句，是为了保持程序结构的完整性。
# pass 不做任何事情，一般用做占位语句，
```
