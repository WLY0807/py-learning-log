```python
# 数据导出
'''
Pandas 提供了丰富的导出功能，可以将 DataFrame 导出为各种常见格式，包括 CSV、Excel、SQL 数据库、JSON 等。
'''
# 导出为 CSV
'''
CSV 是最通用的数据交换格式，导出会受环境默认编码影响，需要注意中文编码问题。
'''
# 示例
import pandas as pd

# 准备数据
df = pd.DataFrame({
    "姓名": ["张三", "李四", "王五", "赵六"],
    "年龄": [25, 30, 28, 35],
    "城市": ["北京", "上海", "广州", "深圳"],
    "薪资": [12000, 15000, 11000, 18000]
})

# 导出为 csv (包含索引)
df.to_csv('test4.csv')
# 不包含索引
df.to_csv('test5.csv',index = False)
# 中文编码处理    UTF-8 编码(Excel 不兼容)
df.to_csv("/mnt/c/Users/18371/Desktop/Pandas 实践集/test-utf-8.csv",encoding = 'utf-8')
# UTF-8 with BOM（Excel 打开不乱码）
df.to_csv("/mnt/c/Users/18371/Desktop/Pandas 实践集/test1-utf-8-sig.csv",encoding = 'utf-8-sig')



# 导出选项详解

# 指定分隔符（默认为 , ）
df.to_csv("/home/wsl2/Pandas-learning-log/测试文档/test1.csv",sep = '\t')  # TSV 格式
'''
TSV = Tab-Separated Values 制表符分隔值
用 Tab 制表符（键盘 Tab 键） 作为每列的分隔符

什么时候优先用 TSV？
数据里大量包含逗号、中英文标点、长文本文案；
自然语言处理、文本数据集（公开数据集大多是 tsv）；
不想处理引号转义、避免列错位问题。
'''
# 不写入表头
df.to_csv('/home/wsl2/Pandas-learning-log/测试文档/output2.csv',header = False)  

# 只导出特定列
df.to_csv('/home/wsl2/Pandas-learning-log/测试文档/output3.csv',columns = ['姓名'])   # 只导出"姓名"列

# 缺失值处理
import numpy as np
df_with_na = pd.DataFrame({
    "A": [1.1314, 2.2596, np.nan, 4.0854],
    "B": ["a", None, "c", "d"]
})
df_with_na.to_csv('/home/wsl2/Pandas-learning-log/测试文档/output4.csv',na_rep = 'NULL',float_format = '%.2f')  # 浮点数精度,保留2位小数
```

```python
# 导出为 Excel
'''
Excel 格式适合人工查看和编辑，但大文件导出较慢。
'''
# 示例
import pandas as pd
df = pd.DataFrame({
    "姓名": ["张三", "李四", "王五"],
    "年龄": [25, 30, 28],
    "城市": ["北京", "上海", "广州"],
    "薪资": [12000, 15000, 11000]
})

# 导出为 Excel（.xlsx 格式，需要 openpyxl）
df.to_excel("/mnt/c/Users/18371/Desktop/Pandas 实践集/test4.xlsx",index = False)  # 不保留索引

# 多 Sheet 导出

# 创建多个 DataFrame
df1 = pd.DataFrame({"A": [1, 2, 3], "B": [4, 5, 6]})
df2 = pd.DataFrame({"C": [7, 8, 9], "D": [10, 11, 12]})

# 导出到同一个 Excel 的不同 Sheet
with pd.ExcelWriter("/mnt/c/Users/18371/Desktop/Pandas 实践集/test5.xlsx") as writer:
    df1.to_excel(writer,sheet_name = 'Sheet1',index = False)
    df2.to_excel(writer,sheet_name = 'Sheet2',index = False)


# 格式化导出
from openpyxl import Workbook
from openpyxl.styles import Font, PatternFill, Alignment

# 使用 ExcelWriter 进行更精细的控制
with pd.ExcelWriter("/mnt/c/Users/18371/Desktop/Pandas 实践集/test6.xlsx") as writer:
    df.to_excel(writer,sheet_name = '员工信息',index = False)

    # 获取工作表
    worksheet = writer.sheets['员工信息']

    # 设置列宽
    worksheet.column_dimensions['A'].width = 15
    worksheet.column_dimensions['B'].width = 10
    worksheet.column_dimensions['C'].width = 12

    # 冻结首行
    worksheet.freeze_panes = 'A2'
```
