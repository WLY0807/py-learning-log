```python
#  数据排序与聚合
'''
数据排序与聚合是数据分析中非常常见且重要的操作，特别是在大数据集中的数据分析时。
排序按特定标准对数据进行排列，而聚合则让我们对数据进行汇总，计算出各种统计量。
Pandas 提供了强大的排序和聚合功能，能够帮助分析人员高效地处理数据。


操作	    方法	                               说明	                                        常用函数/方法
排序	    sort_values(by, ascending)	           根据某列的值进行排序，ascending 控制升降序	    df.sort_values(by='column')
排序	    sort_index(axis)	                   根据行或列的索引进行排序	                    df.sort_index(axis=0)
分组聚合   	groupby(by)	                           按照某列进行分组后，应用聚合函数	            df.groupby('column')
聚合函数	    agg()	                               聚合函数，如 sum()、mean()、count() 等	    df.groupby('column').agg({'value': 'sum'})
多重聚合	    agg([func1, func2])	                   对同一列应用多个聚合函数	                    df.groupby('column').agg({'value': ['mean', 'sum']})
分组后排序	apply(lambda x: x.sort_values(...))	   在分组后进行排序	                            df.groupby('column').apply(lambda x: x.sort_values(...))
透视表	    pivot_table()	                       创建透视表，根据行、列进行数据汇总	
'''
```

```python
# 数据排序（Sorting）
'''
排序是指将数据按某个列的值进行升序或降序排列。Pandas 提供了两种主要的方法来进行排序：sort_values() 和 sort_index()。

排序方法
sort_values()：根据列的值进行排序。
sort_index()：根据行或列的索引进行排序

操作	    方法		    	    	    说明		                	                	        
按值排序	    df.sort_values(by, ascending)	按照指定的列（by）排序，ascending 控制升序或降序，默认为升序
按索引排序	df.sort_index(axis)		        按照行或列的索引排序，axis 控制按行或列排序
'''
# 示例  sort_values()
import pandas as pd

data = {'Name': ['Alice', 'Bob', 'Charlie', 'David'],
        'Age': [25, 30, 35, 40],
        'Salary': [50000, 60000, 70000, 80000]}

df = pd.DataFrame(data)

# 按照 "Age" 列的值进行降序排序
df_sorted = df.sort_values(by = 'Age',ascending = False)
print(df_sorted)
'''
      Name  Age  Salary
3    David   40   80000
2  Charlie   35   70000
1      Bob   30   60000
0    Alice   25   50000
'''

# sort_index() 

# 按照行索引进行排序
df_sorted_by_index = df.sort_index(axis = 0)
print(df_sorted_by_index)
'''
      Name  Age  Salary
0    Alice   25   50000
1      Bob   30   60000
2  Charlie   35   70000
3    David   40   80000
'''
# 按照列索引进行排序
df_sorted_by_index_1 = df.sort_index(axis = 1)
print(df_sorted_by_index_1)
'''
   Age     Name  Salary
0   25    Alice   50000
1   30      Bob   60000
2   35  Charlie   70000
3   40    David   80000

按列名的首字母排序
'''
```

```python
# 数据聚合（Aggregation）
'''
聚合是将数据按某些规则进行汇总，通常是对某些列的数据进行求和、求平均数、求最大值、求最小值等操作。Pandas 提供了groupby()方法来对数据进行分组，然后应用不同的聚合函数。

聚合方法
groupby()：按某些列分组。
聚合函数：如 sum(), mean(), count(), min(), max(), std() 等。

操作	         方法	                               说明
按列分组并聚合	 df.groupby(by).agg()	               按指定列（by）进行分组，agg() 可以传入不同的聚合函数，进行多种操作
多重聚合函数应用	 df.groupby(by).agg([func1, func2])	   可以对同一列应用多个聚合函数，返回多种聚合结果
'''

# 示例  groupby()
import pandas as pd

data = {'Department': ['HR', 'Finance', 'HR', 'IT', 'IT'],
        'Employee': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
        'Salary': [50000, 60000, 55000, 70000, 75000]}
df = pd.DataFrame(data)

# 按照部门分组，并计算每个部门的平均薪资
grouped = df.groupby('Department')['Salary'].mean()
print(grouped)
'''
Department
Finance    60000.0
HR         52500.0
IT         72500.0
Name: Salary, dtype: float64

grouped = df.groupby('Department')['Salary'].mean()
等价于：
# 1. 先分组
group_obj = df.groupby('Department')
# 2. 提取Salary这一列
salary_group = group_obj['Salary']
# 3. 每组求平均值
grouped = salary_group.mean()
'''
# 多重聚合函数应用
# 按照部门分组，并计算每个部门的薪资的平均值和总和
grouped_multiple = df.groupby('Department').agg({
    'Salary':['mean','sum']
})
print(grouped_multiple)
'''
             Salary        
               mean     sum
Department                 
Finance     60000.0   60000
HR          52500.0  105000
IT          72500.0  145000

df.groupby('Department').agg({'Salary':['mean','sum']})
agg( {列名: [统计函数列表]} ) 的作用：
不用单独筛选列，在字典内部指定针对哪一列执行哪些聚合运算，一次性计算多个指标，是分组多统计的标准写法。
'''
# 写法二
grouped_multiple_1 = df.groupby('Department').agg(
    平均薪资 = ('Salary','mean'),
    薪资总和 = ('Salary','sum')
)
print(grouped_multiple_1)
'''
               平均薪资    薪资总和
Department                 
Finance     60000.0   60000
HR          52500.0  105000
IT          72500.0  145000
'''
```

```python
# 分组后的排序
'''
聚合后的数据可以进一步按某列的值进行排序，这样可以找出特定组中最重要的值。

分组后排序
操作	    方法	                                                     说明
分组后排序	df.groupby(by).apply(lambda x: x.sort_values(by='col'))	     在每个分组内部按照某列的值进行排序 
'''
# 示例   

# 按照部门分组后，按各部门内部薪资降序排序
import pandas as pd

data = {'Department': ['HR', 'Finance', 'HR', 'IT', 'IT'],
        'Employee': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
        'Salary': [50000, 60000, 55000, 70000, 75000]}
df = pd.DataFrame(data)

group_sorted = df.groupby('Department').apply(lambda x : x.sort_values(by = 'Salary',ascending = False))
print(group_sorted)
'''
             Department Employee  Salary
Department                              
Finance    1    Finance      Bob   60000
HR         2         HR  Charlie   55000
           0         HR    Alice   50000
IT         4         IT      Eve   75000
           3         IT    David   70000

.apply(lambda x : x.sort_values(by = 'Salary',ascending = False))
apply 内部自动做循环：
取出第 1 个分组（Finance）→ 传给 lambda 的 x，执行排序
取出第 2 个分组（HR）→ 传给 lambda 的 x，执行排序
取出第 3 个分组（IT）→ 传给 lambda 的 x，执行排序

lambda 匿名函数：
只负责一段计算逻辑，没有循环、没有判断分支（只能简单表达式），单独写 lambda 不会循环。

df.apply ()：遍历每一行 / 每一列
df.map ()：遍历 Series 每一个值
df.applymap ()：遍历表格每一个单元格
'''
```

```python
# 透视表
'''
透视表（Pivot Table）是一个特殊的聚合方式，可以让我们通过行、列和聚合函数对数据进行快速汇总，类似于 Excel 中的透视表。
可进行行列交叉二维统计

pivot_table 完整基础语法:
df.pivot_table(
    values=待统计数值列,
    index=作为行的分类列,
    columns=作为列的分类列,
    aggfunc=聚合方式,
    fill_value=空值填充数字
)
参数：
index ： 放在表格行的分类字段，相同值合并一行；对应 groupby 的分组依据。

values ： 要计算的数字列，只能填数值列。

aggfunc ： 支持单个函数、多个函数列表：
aggfunc='mean' 平均值
aggfunc=['sum','mean','max'] 同时计算多个指标

columns（透视表独有，groupby 做不到）：新增列分类，实现二维交叉统计，这是 pivot_table 最大优势。
'''
# 示例

# 使用 pivot_table 计算每个部门的薪资平均值
pivot_table = df.pivot_table(values = 'Salary',index = 'Department',aggfunc = 'mean')
print(pivot_table)
'''
             Salary
Department         
Finance     60000.0
HR          52500.0
IT          72500.0
'''
```
