```python

num = int(input("请输入一个正整数")) #这里如果要想从键盘输入一个数字，不能直接用 num = input("请输入一个正整数")，这样的话会报错：'>' not supported between instances of 'str' and 'int'
if (num>8) :                       #因为字符串 (str) 和 数字 (int) 之间不能用大于号 / 小于号比较，num 代表字符串，而 input 的内容是数字，所以要给 input 函数加上类型
    print("happen")                #Python 中的变量不需要声明。每个变量在使用前都必须赋值，变量赋值以后该变量才会被创建。在 Python 中，变量就是变量，它没有类型，我们所说的类型是变量所指的内存中对象的类型。等号 = 用来给变量赋值。等号左边是变量名，右边是存储在变量中的值。
    print("hello")
elif num<8 :        #python里的判断条件也可以不用加（），直接空格后写表达式和 ： 就行
   print("sorry") 
else : 
   print("happen ",end="")
   print("error")   # print 默认输出是换行的，如果要实现不换行需要在变量末尾加上 end=""
                    # end 是 print 的关键字参数，作用：定义所有内容打印完之后，末尾追加什么字符。默认 end='\n'：\n 是换行符，所以 print 打完内容自动换行。
                    #一个完整的print 封装如下：print(*objects, sep=' ', end='\n', file=sys.stdout, flush=False)
                    # *objects 可变位置参数，可以传入任意多个要打印的数据，多个值用逗号隔开：print(1,2,"abc",True)
                    #sep=' ' 分隔符，默认空格，多个 objects 之间拼接的字符，自定义修改：print(1,2,3, sep="|")  # 1|2|3
                    #file=sys.stdout 输出目标，默认控制台，输出流对象，默认打印到屏幕；也可以传入文件对象，直接写入文件；flush=False 缓冲区刷新，默认关闭
                    #缩进相同的一组语句构成一个代码块，我们称之代码组。
                    #像if、while、def和class这样的复合语句，首行以关键字开始，以冒号( : )结束，该行之后的一行或多行代码构成代码组。
                    #我们将首行及后面的代码组称为一个子句(clause)。end 是 print 的关键字参数，作用：定义所有内容打印完之后，末尾追加什么字符。
                    #默认 end='\n'：\n 是换行符，所以 print 打完内容自动换行。
```
