```python
# Sklearn 基础概念

# 1、数据表示：数据集和特征
'''
数据集（Dataset）
在 scikit-learn 中，数据通常通过两个主要的对象来表示： 特征矩阵 和 目标向量。
特征矩阵（Feature Matrix）：每一行代表一个数据样本，每一列代表一个特征（即输入变量）。
它是一个二维的数组或矩阵，通常使用 NumPy 数组或 pandas DataFrame 来存储。
'''
# 示例
# 有 3 个样本，每个样本有 2 个特征
import numpy as np
X = np.array([[1.0,2.0],[2.0,3.0],[3.0,4.0]])

'''
目标向量（Target Vector）：它表示每个样本的目标（即输出标签），通常是一个一维数组。
'''
# 示例
# 在分类任务中，目标是每个样本的类别标签。对应的目标向量:
y = np.array([0,1,0])   # # 0 类别和 1 类别
'''
特征和标签
特征（Features）：是数据集中用于训练模型的输入变量。在上面的例子中，X 是特征矩阵，包含了所有的输入变量。
标签（Labels）：是机器学习模型的目标输出。在监督学习中，标签是我们希望模型预测的结果。在上面的例子中，y 是标签或目标向量，包含了每个样本的类别。
'''

display('------------------------------------------------------------------------------------------------------------------------------')


'''
数据集分割
在实际应用中，通常需要将数据集分割成训练集和测试集。
scikit-learn 提供了一个方便的函数 train_test_split() 来实现这一点
'''
# 示例
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
'''
以上代码调用 train_test_split 函数，并将结果赋值给四个变量：X_train、X_test、y_train 和 y_test。
X 和 y 是传入 train_test_split 函数的参数，它们分别代表特征数据集和目标变量（标签）。通常 X 是一个二维数组，y 是一个一维数组。
test_size=0.3 参数指定了测试集的大小应该是原始数据集的 30%。这意味着 70% 的数据将被用作训练集，剩下的 30% 将被用作测试集。
random_state=42 参数是一个随机数种子，用于确保每次分割数据集时都能得到相同的结果。这在实验和模型验证中非常有用，因为它确保了结果的可重复性。
'''
```

```python
# 2、模型与算法
'''
监督学习和无监督学习
在 scikit-learn 中，机器学习模型大致分为两大类：监督学习 和 无监督学习。

监督学习（Supervised Learning）：在监督学习中，模型在训练时会利用带标签的数据进行学习，这些标签是我们希望模型预测的结果。
常见的监督学习任务包括分类和回归。
分类（Classification）：将数据点分配到预定的类别中。例如，判断邮件是垃圾邮件还是非垃圾邮件。
回归（Regression）：预测连续值输出。例如，预测房价、气温等。
'''
# 示例

# 从sklearn树模块导入决策树分类类
from sklearn.tree import DecisionTreeClassifier

# 实例化一个默认参数的决策树对象
clf = DecisionTreeClassifier()

# 使用训练集训练决策树
clf.fit(X_train, y_train)

# 在测试集进行预测，输出预测标签
y_pred = clf.predict(X_test)
'''
sklearn.tree：树模型子库，包含决策树分类、回归、绘图工具
DecisionTreeClassifier：决策树分类器，用于离散标签 0/1、多类别任务

clf = DecisionTreeClassifier(
    max_depth=5,          # 树最大深度，限制过拟合，最关键参数
    min_samples_split=2,  # 一个节点至少要有多少样本才允许分裂
    min_samples_leaf=1,   # 叶子节点最少样本数
    criterion="gini",     # 分裂标准：gini基尼系数 / entropy熵
    random_state=42,      # 固定随机种子，保证每次树结构一致
    splitter="best"       # best最优分裂 / random随机分裂
)

.fit()：函数功能：递归生成分裂节点、构建完整决策树

y_pred
输出含义	:真实类别 (0,1,2 代表类别)

高频成员函数
clf.predict_proba(X_test)    # 返回每一类别的预测概率
clf.get_n_leaves()           # 获取叶子节点总数
clf.feature_importances_     # 特征重要性数组，总和=1
'''

'''
无监督学习（Unsupervised Learning）：无监督学习是指没有标签数据，模型仅通过输入数据本身的特征进行学习。
常见的无监督学习任务包括聚类和降维。
聚类（Clustering）：将数据分组，使得同一组中的数据具有相似性。常见的聚类算法包括 K-Means、DBSCAN 等。
降维（Dimensionality Reduction）：减少数据中的特征数量，通常用于数据压缩或可视化。常见的降维方法有 PCA（主成分分析）和 t-SNE（t-分布随机邻域嵌入）等。
'''
# 示例   使用 K-Means 聚类

# 从 sklearn 的聚类模块导入 KMeans（K 均值无监督聚类算法）
from sklearn.cluster import KMeans

# 实例化 K‑Means 对象
kmeans = KMeans(n_clusters = 3)

# 使用训练集训练决策树
kmeans.fit(X_train)

# 在测试集进行预测，输出预测标签
y_pred = kmeans.predict(X_test)
'''
kmeans = KMeans(
    n_clusters=3,          # 最重要：聚类簇的数量 K
    init="k-means++",      # 中心点初始化，默认k‑means++，避免初始点扎堆
    n_init=10,              # 运行10次算法，挑选效果最好的一次中心点
    max_iter=300,           # 单次聚类最大迭代轮数
    random_state=42,        # 固定随机种子，结果复现
    tol=1e-4                 # 中心点变化小于该阈值就提前停止迭代
)

y_pred
输出含义	:簇编号，只是分组标记,没有固定语义

常用属性（训练之后调用）
kmeans.cluster_centers_   # (3,特征数) 数组，3个聚类中心点坐标
kmeans.labels_             # 训练集每一个样本所属簇编号
kmeans.inertia_            # 惯性：所有样本到对应簇中心距离平方和（用来手肘法选K）
'''


'''
预处理与特征工程
在使用 scikit-learn 进行机器学习之前，通常需要对数据进行预处理，这包括以下几项常见任务：
1、标准化（Standardization）：特征的尺度统一，使得每个特征都具有零均值和单位方差。
'''
# 示例

# 从 sklearn 预处理模块导入标准归一化（Z‑score 标准化）工具
from sklearn.preprocessing import StandardScaler

# 创建标准化转换器实例
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X_train)
'''
预处理模块：专门用来做数据缩放、编码、缺失值处理等数据清洗操作

scaler = StandardScaler(
    with_mean=True,      # 是否减去均值，True中心化
    with_std=True         # 是否除以标准差，True缩放至方差为1
)

scaler.fit_transform(X)
等价于：
scaler.fit(X)    # 第一步：计算统计参数
X_scaled = scaler.transform(X)  # 第二步：执行标准化转换

scaler.fit(X)
输入：二维特征矩阵 (样本数, 特征数)
运算：按每一列（每一个特征）分别计算
mean_：该列平均值
var_：方差
scale_ = √方差：标准差
作用：只保存参数，不会改动原始数据 X

scaler.transform(X)套用 Z‑score 标准化公式逐元素计算：
x_new = (x - u)/h
u = fit 得到的特征均值
h = fit 得到的特征标准差
返回处理之后的新数组，原始数组不会被修改

fit_transform()先拟合均值方差，马上转换数据，一步返回缩放后的数据集。
'''


'''
2、归一化（Normalization）：将特征的值缩放到一个固定范围（通常是 0 到 1）。
'''
# 示例

# 从 sklearn 预处理模块导入最大最小值归一化工具
import sklearn.preprocessing import MinMaxScaler

# 实例化归一化器
scaler = MinMaxScaler()

X_normalized = scaler.fit_transform(X)
'''
scaler = MinMaxScaler(
    feature_range=(0, 1),   # 默认缩放到 0～1，也可以设置 (‑1,1)
    clip=False               # 是否把超出区间的值强制截断到边界
)

scaler.min_         # 每个特征的最小值
scaler.max_         # 每个特征的最大值
scaler.data_range_  # max‑min

三个核心函数区分
scaler.fit(X_train)：只用训练集获取最大、最小值
scaler.transform(X_train)：转换训练集
scaler.transform(X_test)：测试集只转换，禁止 fit，防止数据泄露
'''

'''
3、类别变量编码：将类别型数据转换为数值型数据（如 one-hot 编码)
'''
```

```python
# 3、模型评估与验证
'''
在训练机器学习模型之后，必须评估其性能以确保其泛化能力。

交叉验证（Cross-validation）
交叉验证是一种常见的模型评估方法，尤其是在数据量有限的情况下。
通过将数据分成多个子集，每次使用一个子集作为验证集，其余作为训练集，重复多次训练和评估模型，最终计算模型的平均性能。
'''
# 示例

# 从模型选择工具包导入交叉验证评分函数
from sklearn.model_selection import cross_val_score
scores = cross_val_score(clf, X, y, cv = 5)
'''
cross_val_score(
    estimator=clf,        # 机器学习模型
    X=X,                  # 全部特征矩阵
    y=y,                  # 样本标签
    cv=5,                 # 折数；也可以传入划分对象、整数
    scoring="accuracy",   # 评估指标
    n_jobs=-1             # -1 使用电脑全部CPU核心加速运算
)
参数
clf：已经实例化好的模型（决策树、KMeans 不行！K‑Means 无标签 y，不能用这个函数）
X：全部特征矩阵
y：样本标签
cv=5：5‑fold 五折交叉验证

常见评估指标
分类任务的评估指标：

准确率（Accuracy）：预测正确的样本占所有样本的比例。
精确率（Precision）：正类预测中，实际正类的比例。
召回率（Recall）：实际正类中，正确预测的比例。
F1 分数：精确率和召回率的调和平均数。

回归任务的评估指标：
均方误差（MSE）：预测值与真实值的平方差的平均值。
决定系数（R²）：衡量模型对数据变异的解释能力。

五折交叉验证完整流程
将全部数据集随机切分成 5 份互不重叠的数据块
循环 5 次：
第 1 份当作测试集，剩下 4 份当作训练集训练模型
在测试集上计算得分（默认是分类任务准确率 accuracy）
最后返回包含 5 次得分的一维数组 scores
'''
```
