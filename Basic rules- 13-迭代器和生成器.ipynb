```python
# 迭代器
'''
迭代是 Python 最强大的功能之一，是访问集合元素的一种方式。
迭代器是一个可以记住遍历的位置的对象。
迭代器对象从集合的第一个元素开始访问，直到所有的元素被访问完结束。
迭代器只能往前不会后退。
迭代器有两个基本的方法：iter() 和 next()。
字符串，列表或元组对象都可用于创建迭代器

创建一个迭代器：
把一个类作为一个迭代器使用需要在类中实现两个方法 __iter__() 与 __next__() 。
如果你已经了解的面向对象编程，就知道类都有一个构造函数，Python 的构造函数为 __init__(), 它会在对象初始化的时候执行。
__iter__() 方法返回一个特殊的迭代器对象， 这个迭代器对象实现了 __next__() 方法并通过 StopIteration 异常标识迭代的完成。
__next__() 方法（Python 2 里是 next()）会返回下一个迭代器对象。

'''

list = [1,2,3,4]
it = iter(list)     # 创建迭代器对象  
print(next(it))     # 输送迭代器的下一个元素
print(next(it))
print(next(it))
print(next(it))     # 如果输出超过总的迭代数，将会报错
'''
for循环方式
for x in it:
    print(x,end = '  ')
'''
# 创建一个返回数字的迭代器，初始值为 1，逐步递增 1
# 1.定义迭代器类
class MyNumbers:         # 创建名为 MyNumbers 的类，符合 Python 迭代器规范，必须同时拥有 __iter__ 和 __next__ 两个方法。

# 2.__iter__()函数：初始化、返回迭代器自身
    def __iter__(self):  # 定义一个 __iter__() 函数
        self.a = 1       # 其中self.a = 1：给实例绑定属性 a，初始值设为 1
        return self      # return self：规定迭代器必须返回自己本身。 

# 3.__next__()函数：每次调用 next () 执行，返回下一个数值
    def __next__(self):  # 定义一个 __next__ 函数
        x = self.a       # x = self.a：把当前 a 的值临时存给 x；
        self.a += 1      # self.a += 1：a 自增 1，为下一次取值做准备；
        return x         # return x：返回本次拿到的数字。
    
# 4.创建实例 + 获取迭代器
myclass = MyNumbers()   # myclass = MyNumbers()：创建类的实例对象；
myliter = iter(myclass) # iter(myclass)：触发 __iter__() 方法 → self.a = 1，并返回自身赋值给 myiter。
                        # 此时 myiter.a = 1。

print(next(myliter),end = ' ')
print(next(myliter),end = ' ')
print(next(myliter),end = ' ')
print(next(myliter),end = ' ')
'''
for x in myliter:   # 使用for语句会一直迭代下去，无止尽
    print(x,end = ' ')
'''

# StopIteration
'''
StopIteration 异常用于标识迭代的完成，防止出现无限循环的情况，
在 __next__() 方法中我们可以设置在完成指定循环次数后触发 StopIteration 异常来结束迭代。
'''
# 在20次迭代后停止执行
class MyNumbers:
    def __iter__(self):   # 规则：iter() 需要拿到一个迭代器对象，这个对象自带 __next__ 方法；职责：交出 “能持续取值的工具”。
        self.a = 1
        return self       # 智能返回形参 self ,就是当前迭代器实例，所以必须返回它，让后续可以反复调用 __next__。

    def __next__(self):     # 规则：每次 next() 需要拿到本次迭代产出的元素（数字、字符串等业务数据）;职责：交出本次要输出的值，不是交出整个迭代器工具。
        if self.a <= 20:    # 加入限制条件，迭代次数不超过20
            x = self.a
            self.a += 1
            return x         # 返回临时变量的值，即迭代元素（数值 / 字符串）
        else:
            raise StopIteration  # 告诉程序迭代器已经没有下一个元素，迭代结束。
                                 # raise 关键字 主要用来抛出异常对象；StopIteration：迭代器无元素，是内置迭代终止异常。

myclass = MyNumbers()
myiter = iter(myclass)

for x in myiter:
    print(x,end = ' ' )

'''
self.a 完整讲解：对象属性访问语法
1. 语法名称
点语法（成员访问运算符 .）
通用格式：对象.属性名
 .  ：点，专门用来访问对象内部绑定的变量 / 方法
左边：实例对象（这里是 self，代表当前类自己的实例）
右边：属性名称（普通标识符，不能带引号，不是字符串）

2. 拆分 self.a
self：类方法里固定参数，指代当前正在操作的实例对象
.：访问对象内部成员
a：绑定在这个实例身上的属性（实例变量）

3. 为什么要用 self.xxx，而不是直接 a？
如果只写 a = 1，这只是方法内部局部变量，函数运行完就销毁，别的方法拿不到；
self.a 是绑定在实例上的实例属性，整个类所有方法共享。

4. 通俗比喻理解
self：完整工具箱（里面有计数器 current，还自带取数工具__next__）
self.a：工具箱里一张写着数字的纸条

self.xxx 中的xxx规则：
1.硬性语法规则（违反直接报错）
  只能字母、数字、下划线；不能数字开头；不能关键字；无特殊符号空格。
2.软性编码规范（不报错，但推荐遵守）
  普通属性：小写 + 下划线 self.user_name
  内部私有：单下划线 self._temp
  完全隐藏：双下划线 self.__secret
  不要自己写前后双下划线 __xxx__。
'''
```

```python
# 生成器
'''
在 Python 中，使用了 yield 的函数被称为生成器（generator）。
yield 是一个关键字，用于定义生成器函数，生成器函数是一种特殊的函数，可以在迭代过程中逐步产生值，而不是一次性返回所有结果。
跟普通函数不同的是，生成器是一个返回迭代器的函数，只能用于迭代操作，更简单点理解生成器就是一个迭代器。
当在生成器函数中使用 yield 语句时，函数的执行将会暂停，并将 yield 后面的表达式作为当前迭代的值返回。
然后，每次调用生成器的 next() 方法或使用 for 循环进行迭代时，函数会从上次暂停的地方继续执行，直到再次遇到 yield 语句。这样，生成器函数可以逐步产生值，而不需要一次性计算并返回所有结果。
调用一个生成器函数，返回的是一个迭代器对象。
'''

def countdown(n):   # 定义一个生成器函数，使用yield语句逐步产生从 n 到 1 的逆序数字
    while n > 0:
        yield n     # 在每次调用 yield 语句时，函数会返回当前的倒数值，并在下一次调用时从上次暂停的地方继续执行。
        n -= 1      # 自减操作

# 创建生成器对象
generator = countdown(5)

# 通过迭代生成器获取值
print(next(generator))   # 输出：5

# 使用 for 循环迭代生成器
for x in generator:
    print(x,end = ' ')  # 输出：4，3，2，1

print()
'''
只要函数内出现 yield，该函数就不再是普通函数，而是生成器函数。
调用生成器函数不会执行函数体，只会创建一个生成器对象（迭代器）。
yield 作用：
暂停当前函数，把后面的值返回给 next()；
保存函数当前局部变量、执行位置；
下次调用 next()，从上次暂停的 yield 后面继续运行。

和 return 核心区别
return：直接终止函数，销毁局部变量，无法恢复；
yield：暂停、保存状态，可多次恢复执行；
一个函数可以写多个 yield；只能有一个 return。
'''

# 使用 yield 实现斐波那契数列：
import sys                   # 引用系统内置 sys 模块，使用里面 sys.exit()
    
def fibonacci(n):            # 定义生成器函数
    a,b,counter = 0,1,0      # a：本次要产出的斐波那契数；b：下一个斐波那契数；counter：计数，记录当前是第几项
    while True:              # True 为布尔值，此条件一直为真，形成死循环
        if (counter > n):    
            return           # # 函数执行 return 时，生成器会自动抛出 StopIteration，代表迭代结束。
        yield a              # 把当前 a 返回给 next()，打印输出；暂停函数，保存当前所有局部变量状态；下一次 next() 从 yield a 的下一行继续运行。
        a,b, = b,a +b        # 同步更新：新 a = 旧 b，新 b = 旧 a + 旧 b，斐波那契递推公式。
        counter += 1         # 自加操作
f = fibonacci(10)            # f = fibonacci(10)
    
while True:    
    try:                     # 每次调用 next(f) 获取生成器产出的值，同行空格打印；
        print(next(f),end = ' ')
    except StopIteration:    # 当生成器内部 return，抛出 StopIteration，进入 except
        sys.exit()           # 抛出 SystemExit 异常，终止整个 Python 程序。

'''
try-except 异常捕获语法
完整基础语法：
try:
    可能报错的代码块
except 异常类型:
    捕获到异常后执行的代码

执行逻辑：
先运行 try 里所有代码；
如果没报错：except 代码完全跳过，程序正常往下走；
如果抛出的异常 = except 写的异常类型：立刻停止执行 try 剩余代码，跳转执行 except 块；
如果抛出别的异常：不会被捕获，程序直接崩溃报错。
若有多种错误，分开写 except，不同错误做不同处理

使用场景（什么时候必须用 try-except）
1.迭代器 / 生成器用 next 手动取值
迭代完毕会抛StopIteration，用 except 捕获避免程序崩溃，就是你现在的场景；
2.用户输入转换数字、文件读写、网络请求，极易报错；
3.不确定代码是否会报错，不想程序直接终止；
4.需要区分错误类型，给用户友好提示，而不是一堆红色报错。

sys.exit () 模块退出函数
Jupyter 环境会提示异常日志：An exception has occurred, use %tb to see the full traceback.
                          SystemExit

yield 是 Python 解释器底层内置封装好的语法机制，不是普通函数，解释器对带 yield 的函数做特殊处理：
普通函数：调用时从头执行到底，局部变量执行完销毁；
含 yield 的函数 → 自动封装为生成器（Generator），自带完整迭代器协议：内置 __iter__()、__next__()，自动处理状态保存、暂停、StopIteration，不用你手动写类迭代器。        
'''
```
