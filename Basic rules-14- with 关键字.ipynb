```python
# with 关键字
'''
with 关键字为我们提供了一种优雅的方式来处理文件操作、数据库连接等需要明确释放资源的场景。
with 是 Python 中的一个关键字，用于上下文管理协议（Context Management Protocol）。它简化了资源管理代码，特别是那些需要明确释放或清理的资源（如文件、网络连接、数据库连接等）。
'''

# 为什么需要 with 语句
'''
一个典型的文件操作示例：
file = open ('example.txt','r')
try:
    content = file.read()
    # 处理文件内容
finally:
    file.close()

这种写法存在几个问题：
容易忘记关闭资源：如果没有 try-finally 块，可能会忘记调用 close()
代码冗长：简单的文件操作需要多行代码
异常处理复杂：需要手动处理可能出现的异常
'''

# with 语句的基本语法
'''
基础用法
with 语句的基本形式如下：
with expression [as variable]:
    # 代码块
    
expression 返回一个支持上下文管理协议的对象
as variable 是可选的，用于将表达式结果赋值给变量
代码块执行完毕后，自动调用清理方法
'''
# 最常见的 with 语句应用是文件操作：
with open('hello world.ipynb', 'r') as file:   # 标准格式：open(文件路径, mode='r', encoding="utf-8")
                                                 # with ... as file 上下文管理器
    content = file.read()    # read()：一次性读取文件全部文本，拼接成一整个长字符串返回；
    print(content)

'''
open('Basic rules-4.ipynb', 'r') :

'Basic rules-4.ipynb'：目标文件路径；
mode='r'：只读模式（read），只能读取，不能写入；
返回文件句柄对象（IO 流对象），拥有 read() / readline() / close() 等方法。
常用文件 mode 模式:
模式	作用
r	    只读，文件不存在报错
w	    只写，不存在则新建，存在直接清空覆盖
a	    追加写入，光标在文件末尾
r+	    可读可写

with ... as file  :

with 是 Python 上下文管理器协议语法。
作用：自动管理资源生命周期。
as file：把 open() 返回的文件对象绑定到变量 file；
缩进内：文件处于打开状态，可读写；
当代码退出 with 缩进块（正常执行完毕 / 中途报错），自动执行 file.close() 关闭文件。

能放在 with xxx as file 后的对象，必须实现两个魔法方法：
__enter__(self)
进入 with 代码块时自动执行，返回值赋值给 as 后的变量；
文件场景：打开文件，返回文件对象。
__exit__(self, exc_type, exc_val, exc_tb)
退出代码块时自动执行，无论是否报错都会运行；
文件场景：内部封装 self.close()，完成资源释放。
'''
```

```python
# 实际应用场景
# 1. 文件操作

# 同时打开多个文件
with open("测试文本1.txt",'r') as infile, open("测试文本2.txt",'w') as outfile:
    content = infile.read()
    outfile.write(content.upper())

# 此段程序 读取测试文本 1全部内容，英文转大写后，覆盖 / 新建写入到测试文本 2，全程自动开关文件，控制台无任何打印输出。
'''
content.upper()：字符串方法，所有英文字母转大写，中文、数字、符号不变；

open(..., 'w') 覆盖模式：
若「测试文本 2.txt」不存在 → 新建文件；
若已存在 → 直接清空原有全部内容，再写入新数据；
write() 只写入文件，控制台无任何打印输出。
'''

# 2. 数据库连接
import sqlite3   # 导入Python内置SQLite数据库模块，无需额外安装，轻量文件型数据库
# 1. 建立数据库连接 + 上下文管理器自动管理连接
with sqlite3.connect('database.db') as conn:
    # 2. 创建游标cursor
    cursor = conn.cursor()
    # 3. 建 users数据表
    cursor.execute('''
    CREATE TABLE IF NOT EXISTS users (
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        name TEXT,
        age INTEGER
    )
    ''')
    # 4. DML语句：插入数据（新增一条用户记录）
    cursor.execute("INSERT INTO users(name, age) VALUES (?, ?)", ("雷佳", 22))
    # 5. DQL语句：查询整张表所有数据
    cursor.execute('SELECT * FROM users')
    # 6. 取出查询结果
    results = cursor.fetchall()
    print(results)
'''
sqlite3.connect('database.db')：打开/创建database.db数据库文件
with 上下文管理器：代码块结束自动commit提交事务、close关闭连接

cursor = conn.cursor()
所有SQL语句（建表、增删改查）都必须通过游标执行，是数据库交互载体

SQLite 官方标准语法，字段说明：
CREATE TABLE IF NOT EXISTS：防止重复运行代码报“表已存在”错误
id INTEGER PRIMARY KEY AUTOINCREMENT
- INTEGER：整型
- PRIMARY KEY：主键，每条数据唯一标识，不可重复、不能为空
- AUTOINCREMENT：SQLite 专用关键字，自增，新增数据自动分配1,2,3...无需手动传值
name TEXT：文本类型，存储姓名（字符串）
age INTEGER：整型，存储年龄

cursor.execute("INSERT INTO users(name, age) VALUES (?, ?)", ("雷佳", 22))
INSERT INTO 表名(字段1,字段2) VALUES(?, ?)
INSERT
含义：插入新增数据
属于 DML（数据操作语言），专门用来往表里添加新记录。
INTO
搭配 INSERT 使用，固定组合 INSERT INTO，表示「插入到哪一张表」。
VALUES
含义：值列表;? 是sqlite3专属占位符，防止SQL注入，安全写法
后面元组("雷佳", 22)会依次替换两个?，对应name、age

cursor.execute('SELECT * FROM users')
SELECT * FROM users  ;  * 代表查询表内全部字段；FROM users 指定查询users表
SELECT
作用：查询、读取表中的数据
归类：DQL 查询语言，专门用来获取数据
FROM
固定搭配 SELECT xxx FROM 表名，含义：数据来自哪一张表
FROM users = 数据来源是 users 表

fetchall()：一次性获取本次查询匹配的**全部行数据**
返回格式：列表嵌套元组，一条记录 = 一个元组
'''
```

```python
# 3. 线程锁

# 导入多线程模块，用于创建、管理线程与锁
import threading       # 导入 Python 内置多线程库，提供线程创建、锁、线程同步工具。

# 创建互斥锁对象
lock = threading.Lock()

# with上下文管理锁：自动加锁、释放锁
with lock:
    # 临界区：同一时间只能有一个线程执行这里的代码
    print('这段代码是线程安全的')

'''
threading.Lock()
创建互斥锁（基础锁），核心特性：
锁只有两种状态：未锁定 / 已锁定
同一时刻只能有 1 个线程拿到锁；其他线程会阻塞等待，直到锁被释放
作用：保护临界区代码，防止多线程同时修改共享变量引发数据错乱
底层两个核心方法（with 会自动调用）：
lock.acquire()：申请加锁，拿不到就阻塞等待
lock.release()：释放锁，其他线程可竞争获取

with lock  等价于：
lock.acquire()   # 自动加锁
try:
    print("这段代码是线程安全的")
finally:
    lock.release() # 无论是否报错，自动释放锁

不加锁的问题:
多线程同时 print，输出会互相穿插、乱序；操作全局数字会出现计算错误（数据竞争）。
Lock 不可重复加锁:
同一个线程连续 acquire() 会死锁；需要重复上锁用 threading.RLock() 可重入锁。
锁的作用：线程同步，保证共享资源操作原子性。
'''
```

```python
# 4.临时修改系统状态

# 导入decimal库，用于专门解决普通浮点数 float 精度丢失问题。
import decimal
# 单独导入Decimal类，用来创建高精度小数对象
from decimal import Decimal

# 获取当前全局计算上下文的精度并打印，默认全局prec=28
print("全局默认精度：", decimal.getcontext().prec)  # 全局默认精度： 28

# 用字符串构造Decimal避免float精度丢失，执行 1÷7
num = Decimal('1') / Decimal('7')
print("默认精度结果：", num)    # 默认精度结果： 0.1428571428571428571428571429

# 开启局部上下文管理器，临时隔离精度配置
with decimal.localcontext() as ctx:
    # 修改局部上下文有效数字精度为42位，仅块内生效
    ctx.prec = 42
    # 同一条除法运算，使用42位有效数字计算
    num_high = Decimal('1') / Decimal('7')
    print("with内42位高精度：", num_high)   # with内42位高精度： 0.142857142857142857142857142857142857142857

# 退出with，精度复原
print("退出后精度：", decimal.getcontext().prec)  # 退出后精度： 28

'''
decimal.getcontext().prec
读取全局上下文，全局所有 Decimal 运算默认使用这里的 prec、舍入方式等配置，出厂默认精度 28。

decimal.localcontext()
生成独立的局部上下文，支持 with 上下文协议：
进入 with：保存旧全局配置，切换为局部配置；
退出 with（正常结束 / 报错）：自动还原全局配置，不会污染外部代码。

ctx.prec
prec = precision，代表有效数字总位数（不是小数点后位数）；数值越大，小数保留位数越多，精度越高。

Decimal('1')
必须传入字符串创建高精度小数，如果写 Decimal(1.0) 会先经过 float 二进制失真，失去高精度意义。
num_high = Decimal('1') / Decimal('7')
同样 1÷7，现在运算规则是局部 42 位精度，输出 42 位有效数字:0.142857142857142857142857142857142857142857

from 模块名 import 类/函数/变量
作用：只从指定模块里拿出你需要的对象，直接加载到当前代码的作用域，后续使用不用加模块前缀。
'''
```

```python
# 创建自定义的上下文管理器

# 类实现方式
# 我们可以通过实现 __enter__ 和 __exit__ 方法创建自定义的上下文管理器：
import time

class Timer:     # 定义一个计时器类；必须同时实现两个魔法方法：__enter__()，__exit__()
    def __enter__(self):    # 进入 with 代码块时自动执行
        # 进入上下文，记录开始时间
        self.start = time.time()    # 获取当前开始时间
        return self    # 把当前Timer实例返回，赋值给 with ... as t 中的变量t

    def __exit__(self,exc_type,exc_val,exc_tb):  # 离开 with 代码块时自动执行（无论正常结束、报错、return 都会执行）
        # 退出上下文，计算耗时
        self.end = time.time()   # 获取当前结束时间
        self.cost = self.end - self.start    # 保存耗时供外部读取
        print(f"耗时：{self.cost : .2f}秒")
        return False     # return False 表示不屏蔽代码块内异常，原样抛出

# 使用实例
with Timer() as t:
    s = 0
    for i in range(2000000):
        s += i
print(f"外部读取耗时：{t.cost:.3f} 秒")

'''
self.start = time.time()  
self代表Timer实例对象，给实例新增属性start
time.time() 获取系统时间戳：从1970-01-01到现在的总秒数

def __exit__(self, exc_type, exc_val, exc_tb):
三个参数含义（代码块有无异常，都会传参），且一个不能少
exc_type----exception type 异常类型
代码块无报错：None
代码块抛出异常：异常类型对象，如 ValueError、IndexError
exc_val-----exception value 异常实例 / 异常信息
无报错：None
有异常：异常实例，如 ValueError("数值错误")
exc_tb----exc_tb	exception traceback 异常堆栈追踪对象
无报错：None
有异常：异常堆栈跟踪对象，可打印报错行号、报错文件

return 返回值作用（True / False 天差地别）
return False / 不写 return（默认 None 等价 False）：不捕获异常
with 内部代码报错后，异常继续向上抛出，程序崩溃，正常调试。
return True：吞噬所有异常
代码块里哪怕报错，外部完全看不到，程序不会中断，隐藏 bug，极少使用。一般不推荐使用

self.cost ：.2f  
.2f：保留 2 位浮点小数打印；.2表示时间精度；f 表示数据类型
变量 ：.精度 类型  ，中间的 ： 是分隔符，不能少
'''

# 使用contextlib 模块
# Python 的 contextlib 模块提供了更简单的方式来创建上下文管理器：
from contextlib import contextmanager

@contextmanager
def tag(name):
    # 第一段：进入with块之前执行（对应类写法的 __enter__）
    print(f"<{name}>")
    
    # 暂停点，交出执行权给with内部代码
    yield

    # 第二段：离开with块时执行（对应类写法的 __exit__）
    print(f"/<{name}>")

# 使用实例
with tag("h1"):
    print("这是一个标题")

'''
from contextlib import contextmanager
contextlib：Python 内置工具库，专门简化上下文管理器编写
contextmanager：装饰器，作用：把生成器函数转换成支持 with 的上下文对象

@contextmanager
def tag(name):
@装饰器 等价于 函数名 = 装饰器(原函数)；
作用是把原函数丢给装饰器加工，用加工后的新版本覆盖原函数，全程不用修改原函数内部代码。
contextmanager 是系统提供的装饰器函数；
把我们写的原始 tag 生成器，作为参数传给 contextmanager；
contextmanager 内部自动封装了 __enter__ / __exit__ 魔法方法
@contextmanager 等价于这一句核心代码：
tag = contextmanager(tag)
'''
```
