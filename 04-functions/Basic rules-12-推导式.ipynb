```python
# python 推导式
'''
Python 推导式是一种独特的数据处理方式，可以从一个数据序列构建另一个新的数据序列的结构体。
Python 推导式是一种强大且简洁的语法，适用于生成列表、字典、集合和生成器。
在使用推导式时，需要注意可读性，尽量保持表达式简洁，以免影响代码的可读性和可维护性。
Python 支持各种数据结构的推导式：
列表(list)推导式
字典(dict)推导式
集合(set)推导式
元组(tuple)推导式
'''
```

```python
# 列表推导式
'''
列表推导式格式为：
[表达式 for 变量 in 列表] 
[out_exp_res for out_exp in input_list]  # 我的理解就是变量结合了一个for循环

out_exp_res：对变量运算后的结果，存入新列表
out_exp：循环变量，每次取出序列一个元素
input_list：原序列（列表、字符串、range 等）

或者 

[表达式 for 变量 in 列表 if 条件]   # 我的理解就是表达式结合了一个for循环，并且这个for循环有额外条件
[out_exp_res for out_exp in input_list if condition]
out_exp_res：列表生成元素表达式，可以是有返回值的函数。
for out_exp in input_list：迭代 input_list 将 out_exp 传入到 out_exp_res 表达式中。
if condition：条件语句，只有满足 condition 的元素，才会执行表达式并放进新列表。
'''

# 过滤掉长度小于或等于3的字符串列表，并将剩下的转换成大写字母：
names = ["Kobe","James","Curry","Irving","Lyy"]
new_names = [name.upper() for name in names if len(name) > 3]
print(new_names)   # ['KOBE', 'JAMES', 'CURRY', 'IRVING']
# upper() 是字符串内置方法，只给字符串变量调用，作用：把字符串里所有小写英文字母 → 大写字母。
# 语法：字符串.upper()

# 计算 30 以内可以被 3 整除的整数：
multiples = [i for i in range(30) if i % 3 == 0]
print(multiples)  # [0, 3, 6, 9, 12, 15, 18, 21, 24, 27]
```

```python
# 字典推导式
'''
{ key_expr: value_expr for value in collection }

collection：可迭代对象（列表、元组、range、旧字典等）
value：遍历时取出的单个元素
key_expr：表达式，生成字典的键
value_expr：表达式，生成字典的值


或

{ key_expr: value_expr for value in collection if condition }

if condition：可选，只保留满足条件的键值对
'''

# 使用字符串及其长度创建字典：
listdemo = ["google","baidu","edge"]
# 将列表中各字符串值为键，各字符串的长度为值，组成键值对
newdict = {key : len(key) for key in listdemo}
print(newdict)  # {'google': 6, 'baidu': 5, 'edge': 4}

# 提供几个数字，以合格的数字为键，数字的平方为值来创建字典：
dic = {x : x ** 2 for x in (2,4,6,3,5,7) if x % 2 == 1}
print(dic)  # {3: 9, 5: 25, 7: 49}
```

```python
# 集合推导式
'''
集合推导式基本格式：
{ expression for item in Sequence }

Sequence：任意可迭代对象（列表、字符串、range、元组等）
item：遍历每一个元素的临时变量
expression：对 item 处理后的表达式，作为集合里的元素

或
{ expression for item in Sequence if conditional }

只有满足 conditional 条件的元素，才会经过 expression 运算存入集合。
'''

# 计算数字 1,2,3 的平方数：
set_2 = {x ** 2 for x in range(1,5)}
print(set_2)  # {16, 1, 4, 9}

# 判断不是 abc 的字母并输出：
set_num = {x for x in "abdndbcnng" if x not in "abc"}
print(set_num)  # {'n', 'g', 'd'}
```

```python
# 元组推导式
'''
元组推导式可以利用 range 区间、元组、列表、字典和集合等数据类型，快速生成一个满足指定需求的元组。
元组推导式基本格式：
(expression for item in Sequence )
或
(expression for item in Sequence if conditional )
'''

# 生成一个包含数字 1~9 的元组：
set_zu = (x for x in range(1 ,10))
print(set_zu)  # <generator object <genexpr> at 0x7431295392a0>
               # 返回的是生成器对象
tuple(set_zu)  # (1, 2, 3, 4, 5, 6, 7, 8, 9)
               # 使用 tuple() 函数，可以直接将生成器对象转换成元组
```
