```python
# 虚拟环境
'''
虚拟环境是一个独立的 Python 运行空间，拥有自己的解释器、安装包和配置，与系统全局环境完全隔离。
Python 虚拟环境（Virtual Environment）是一个独立的 Python 运行环境，它允许你在同一台机器上为不同的项目创建隔离的 Python 环境。
不同项目可以在各自的虚拟环境中并行运行，互不干扰。
虚拟环境让每个项目拥有独立的依赖空间，彻底消除版本冲突

每个虚拟环境都有自己的：
Python 解释器
安装的包/库
环境变量

为什么需要虚拟环境:
    项目隔离：不同项目可使用不同版本的 Python 和第三方库
    避免污染：安装的包只影响当前环境，不污染全局 Python
    依赖可控：通过 requirements.txt 精确记录和复现环境
    安全测试：可以放心升级或试用新包，不影响其他项目
场景举例：
    项目 A 需要 Django 3.2 版本
    项目 B 需要 Django 4.0 版本
    如果在系统全局安装，两个版本会冲突

虚拟环境工具
工具名称  	    类型	        Python版本支持	安装方式	                     特点	                    适用场景
venv	        内置模块     	≥ 3.3	        无需安装，内置	轻量级、     官方推荐、使用简单	        通用开发、日常项目
virtualenv	    第三方工具	    2.x 和 3.x	    pip install virtualenv	    功能丰富、兼容多版本	        需要兼容旧版本或高级功能
conda	        Anaconda自带	2.x 和 3.x	    随 Anaconda/Miniconda 安装	跨语言包管理、数据科学生态	数据科学、机器学习项目
'''
```

```python
# 创建虚拟环境（全部操作在 WSL2/Linux 命令行运行）
'''
conda 环境

完整标准操作全流程（8 步闭环）
第 1 步：检查当前 conda 环境状态
# 查看所有已创建的环境列表，带*的是当前激活的环境
conda env list
# 简写版
conda info --envs

第 2 步：创建全新虚拟环境
标准格式
conda create -n 环境名 python=指定版本号
常用示例（AI 通用稳定版）
# 创建名为 ai_exp 的环境，Python版本为3.10
conda create -n ai_exp python=3.10
执行后出现确认提示，输入 y 回车，等待环境创建完成
可选参数：末尾加 -y 可跳过确认，直接创建：conda create -n ai_exp python=3.10 -y

第 3 步：激活目标环境
conda activate 环境名
# 示例
conda activate ai_exp
激活成功标志：终端行首出现 (环境名) 前缀，此时所有 python/pip/conda 命令仅作用于当前环境

第 4 步：安装项目依赖包
优先使用 conda 安装（数值 / 深度学习库首选，自动处理底层依赖）
# 标准格式
conda install 包名
# 指定版本
conda install 包名=版本号
# 批量安装多个包
conda install numpy pandas scikit-learn
补充使用 pip 安装（conda 源没有的小众 Python 包）
# 标准格式
pip install 包名
# 指定版本
pip install 包名==版本号

第 5 步：开发运行代码
# 运行Python脚本
python 脚本名.py
# 进入Python交互式终端
python
# 启动Jupyter（AI实验常用，需先conda install jupyter）
jupyter lab

第 6 步：导出 / 备份实验环境（课题复现必备）
# 导出完整环境配置（含Python版本、所有依赖、底层库），生成environment.yml文件
conda env export > environment.yml

第 7 步：退出当前虚拟环境
conda deactivate
执行后终端前缀的环境标识消失，回到系统默认 /base 环境

第 8 步：删除不需要的环境
# 必须先退出目标环境，再执行删除
conda remove -n 环境名 --all
# 示例
conda remove -n ai_exp --all



高频命令速查表
1. 环境管理类
操作	                    命令
查看所有环境	                conda env list
创建环境（指定 Python 版本）	conda create -n 环境名 python=3.10
激活环境	                    conda activate 环境名
退出环境	                    conda deactivate
重命名环境	                conda rename -n 旧名 新名
彻底删除环境	                conda remove -n 环境名 --all
克隆复制环境	                conda create -n 新环境名 --clone 旧环境名

2. 包管理类
操作	                 命令
查看当前环境已安装的包	 conda list
conda 安装包	         conda install 包名
conda 安装指定版本包	     conda install 包名=2.0.0
conda 卸载包	         conda remove 包名
conda 更新包	         conda update 包名
搜索源内可用的包版本	     conda search 包名

3. 配置类
操作	                    命令
查看当前 conda 配置	        conda config --show
添加清华镜像源（国内加速）	conda config --add channels https://mirrors.tuna.tsinghua.edu.cn/anaconda/pkgs/main/
开启安装时显示源地址	        conda config --set show_channel_urls yes
恢复默认源	                conda config --remove-key channels


强制避坑规范
绝对不要在 base 环境内装项目依赖、写代码，所有实验必须新建独立环境
禁止 conda 和 venv 混用，全程用 conda 就不要再执行python -m venv创建虚拟环境，避免双层嵌套导致包安装错乱
深度学习 / 数值计算库优先用 conda install，pip 安装容易出现 GPU 不可用、底层库缺失问题
激活环境后再装包 / 运行代码，未激活状态下装包会安装到 base 环境，造成环境污染
每个独立课题 / 实验单独创建一个环境，不要所有项目共用一个环境，避免版本冲突

'''
```
