```python
# 数据清洗
'''
数据清洗是对一些没有用的数据进行处理的过程。
很多数据集存在数据缺失、数据格式错误、错误数据或重复数据的情况，如果要使数据分析更加准确，就需要对这些没有用的数据进行处理。

数据清洗与预处理的常见步骤：
1.缺失值处理：识别并填补缺失值，或删除含缺失值的行/列。
2.重复数据处理：检查并删除重复数据，确保每条数据唯一。
3.异常值处理：识别并处理异常值，如极端值、错误值。
4.数据格式转换：转换数据类型或进行单位转换，如日期格式转换。
5.标准化与归一化：对数值型数据进行标准化（如 Z-score）或归一化（如 Min-Max）。
6.类别数据编码：将类别变量转换为数值形式，常见方法包括 One-Hot 编码和标签编码。
7.文本处理：对文本数据进行清洗，如去除停用词、词干化、分词等。
8.数据抽样：从数据集中抽取样本，或通过过采样/欠采样处理类别不平衡。
9.特征工程：创建新特征、删除不相关特征、选择重要特征等。
'''
```

```python
# Pandas 清洗空值
'''
如果要删除包含空字段的行，可以使用 dropna() 方法，语法格式如下：

DataFrame.dropna(axis=0, how='any', thresh=None, subset=None, inplace=False)
参数说明：
axis：默认为 0，表示逢空值剔除整行，如果设置参数 axis＝1 表示逢空值去掉整列。
how：默认为 'any' 如果一行（或一列）里任何一个数据有出现 NA 就去掉整行，如果设置 how='all' 一行（或列）都是 NA 才去掉这整行。
thresh：设置需要多少非空值的数据才可以保留下来的。
subset：设置想要检查的列。如果是多个列，可以使用列名的 list 作为参数。
inplace：如果设置 True，将计算得到的值直接覆盖之前的值并返回 None，修改的是源数据。

可以通过 isnull() 判断各个单元格是否为空。
'''
# 示例
import pandas as pd

site = "/mnt/c/Users/18371/Desktop/Pandas 实践集/property-data.csv"
df = pd.read_csv(site)
'''
上 csv 文件包含了四种空数据：
n/a    NA     —     na
'''

# 单独提取 NUM_BEDROOMS 这一列，打印全部数值
print(df['NUM_BEDROOMS'])
'''
0      3
1      3
2    NaN
3      1
4      3
5    NaN
6      2
7      1
8     na
Name: NUM_BEDROOMS, dtype: object
'''

# .isnull() 判断每个单元格是否是空值 NaN
# 返回布尔Series：True=是空值，False=有数据
print(df['NUM_BEDROOMS'].isnull())
'''
0    False
1    False
2     True
3    False
4    False
5     True
6    False
7    False
8    False
Name: NUM_BEDROOMS, dtype: bool

以上例子中可以看到 Pandas 把 NAN 当作空数据，na 不是空数据，不符合要求
'''

# 指定空数据类型
missing_values = ['n/a','na','--']
df = pd.read_csv(site,na_values = missing_values)

print (df['NUM_BEDROOMS'].isnull())
'''
0    False
1    False
2     True
3    False
4    False
5     True
6    False
7    False
8     True
Name: NUM_BEDROOMS, dtype: bool

可以看到这次 na 也被 True 了
'''

# 删除包含空数据的行
df = pd.read_csv(site,na_values = missing_values)

# df.dropna()：删除**任意一列存在空值NaN**的整行，返回新DataFrame（原df不变）
new_df = df.dropna()

# to_string()：打印完整全部数据，不会省略中间行
print(new_df.to_string())
'''
           PID  ST_NUM ST_NAME OWN_OCCUPIED  NUM_BEDROOMS NUM_BATH   SQ_FT
0  100001000.0   104.0  PUTNAM            Y           3.0        1  1000.0

默认情况下，dropna() 方法返回一个新的 DataFrame，不会修改源数据
'''
# 修改源数据 DataFrame, 使用 inplace = True 参数
site_1 =  "/mnt/c/Users/18371/Desktop/Pandas 实践集/property-data - 副本.csv"
df1 = pd.read_csv(site_1,na_values = missing_values)

df1.dropna(inplace = True)
print(df1.to_string())
'''
           PID  ST_NUM ST_NAME OWN_OCCUPIED  NUM_BEDROOMS NUM_BATH   SQ_FT
0  100001000.0   104.0  PUTNAM            Y           3.0        1  1000.0

inplace=True 含义：就地修改当前 df 对象，不返回新表格
执行后，你代码里的df直接删除所有含空值的行，后续所有操作用的都是清洗后的数据
改变的并不是硬盘上的原始数据
'''

# 移除指定列有空值的行
# 移除 ST_NUM 列中字段值为空的行
df = pd.read_csv(site)

df.dropna(subset = ['ST_NUM'],inplace = True)

print(df.to_string())
'''
           PID  ST_NUM    ST_NAME OWN_OCCUPIED NUM_BEDROOMS NUM_BATH SQ_FT
0  100001000.0   104.0     PUTNAM            Y            3        1  1000
1  100002000.0   197.0  LEXINGTON            N            3      1.5    --
3  100004000.0   201.0   BERKELEY           12            1      NaN   700
4          NaN   203.0   BERKELEY            Y            3        2  1600
5  100006000.0   207.0   BERKELEY            Y          NaN        1   800
7  100008000.0   213.0    TREMONT            Y            1        1   NaN
8  100009000.0   215.0    TREMONT            Y           na        2  1800

这里依旧只识别默认的 Na 为空值
'''

# 用 fillna() 方法来替换一些空字段
# 使用 12345 替换空字段
df = pd.read_csv(site)

df.fillna(12345)

print(df.to_string())
'''
           PID   ST_NUM     ST_NAME OWN_OCCUPIED NUM_BEDROOMS NUM_BATH  SQ_FT
0  100001000.0    104.0      PUTNAM            Y            3        1   1000
1  100002000.0    197.0   LEXINGTON            N            3      1.5     --
2  100003000.0  12345.0   LEXINGTON            N        12345        1    850
3  100004000.0    201.0    BERKELEY           12            1    12345    700
4      12345.0    203.0    BERKELEY            Y            3        2   1600
5  100006000.0    207.0    BERKELEY            Y        12345        1    800
6  100007000.0  12345.0  WASHINGTON        12345            2   HURLEY    950
7  100008000.0    213.0     TREMONT            Y            1        1  12345
8  100009000.0    215.0     TREMONT            Y           na        2   1800
'''

# 指定某一个列来替换数据
# 使用 12345 替换 PID 为空数据
df['PID'].fillna(12345,inplace = True)   # 链式法则
print(df.to_string())
'''
           PID  ST_NUM     ST_NAME OWN_OCCUPIED NUM_BEDROOMS NUM_BATH SQ_FT
0  100001000.0   104.0      PUTNAM            Y            3        1  1000
1  100002000.0   197.0   LEXINGTON            N            3      1.5    --
2  100003000.0     NaN   LEXINGTON            N          NaN        1   850
3  100004000.0   201.0    BERKELEY           12            1      NaN   700
4      12345.0   203.0    BERKELEY            Y            3        2  1600
5  100006000.0   207.0    BERKELEY            Y          NaN        1   800
6  100007000.0     NaN  WASHINGTON          NaN            2   HURLEY   950
7  100008000.0   213.0     TREMONT            Y            1        1   NaN
8  100009000.0   215.0     TREMONT            Y           na        2  1800
'''


# 替换空单元格的常用方法是计算列的均值、中位数值或众数
'''
Pandas使用 mean()、median() 和 mode() 方法计算列的均值（所有值加起来的平均值）、
中位数值（排序后排在中间的数）和众数（出现频率最高的数）。
'''
# 示例   使用 mean() 方法计算列的均值并替换空单元格
x = df['ST_NUM'].mean()
df['ST_NUM'].fillna(x,inplace = True)

print(df.to_string())
'''
           PID      ST_NUM     ST_NAME OWN_OCCUPIED NUM_BEDROOMS NUM_BATH SQ_FT
0  100001000.0  104.000000      PUTNAM            Y            3        1  1000
1  100002000.0  197.000000   LEXINGTON            N            3      1.5    --
2  100003000.0  191.428571   LEXINGTON            N          NaN        1   850
3  100004000.0  201.000000    BERKELEY           12            1      NaN   700
4      12345.0  203.000000    BERKELEY            Y            3        2  1600
5  100006000.0  207.000000    BERKELEY            Y          NaN        1   800
6  100007000.0  191.428571  WASHINGTON          NaN            2   HURLEY   950
7  100008000.0  213.000000     TREMONT            Y            1        1   NaN
8  100009000.0  215.000000     TREMONT            Y           na        2  1800

ST_NUM列中的空值被替换为均值 191.428571
'''

# 使用 median() 方法计算列的众数并替换空单元格
y = df['PID'].median()
df['PID'].fillna(y,inplace = True)

print(df.to_string())
'''
           PID      ST_NUM     ST_NAME OWN_OCCUPIED NUM_BEDROOMS NUM_BATH SQ_FT
0  100001000.0  104.000000      PUTNAM            Y            3        1  1000
1  100002000.0  197.000000   LEXINGTON            N            3      1.5    --
2  100003000.0  191.428571   LEXINGTON            N          NaN        1   850
3  100004000.0  201.000000    BERKELEY           12            1      NaN   700
4  100005000.0  203.000000    BERKELEY            Y            3        2  1600
5  100006000.0  207.000000    BERKELEY            Y          NaN        1   800
6  100007000.0  191.428571  WASHINGTON          NaN            2   HURLEY   950
7  100008000.0  213.000000     TREMONT            Y            1        1   NaN
8  100009000.0  215.000000     TREMONT            Y           na        2  1800

这里是将 12345 去掉以后的中位数
PID 列中的空值被替换为中位数 100005000.0
'''

# 使用 mode() 方法计算列的众数并替换空单元格
df5 = pd.read_csv(site,na_values = missing_values)
z = df5['NUM_BEDROOMS'].median()
df5['NUM_BEDROOMS'].fillna(z,inplace = True)

print(df5.to_string())
'''
           PID  ST_NUM     ST_NAME OWN_OCCUPIED  NUM_BEDROOMS NUM_BATH   SQ_FT
0  100001000.0   104.0      PUTNAM            Y           3.0        1  1000.0
1  100002000.0   197.0   LEXINGTON            N           3.0      1.5     NaN
2  100003000.0     NaN   LEXINGTON            N           2.5        1   850.0
3  100004000.0   201.0    BERKELEY           12           1.0      NaN   700.0
4          NaN   203.0    BERKELEY            Y           3.0        2  1600.0
5  100006000.0   207.0    BERKELEY            Y           2.5        1   800.0
6  100007000.0     NaN  WASHINGTON          NaN           2.0   HURLEY   950.0
7  100008000.0   213.0     TREMONT            Y           1.0        1     NaN
8  100009000.0   215.0     TREMONT            Y           2.5        2  1800.0

原数据中，有na,Nan 两种空值，但默认只有 Nan 为空值，na 不被识别，则不能对这列的数据进行数学运算，所以将 na 也定义为了空值
NUM_BEDROOMS  中的空值被替代为众数 3
'''
```

```python
# 清洗格式错误数据
'''
数据格式错误的单元格会使数据分析变得困难，甚至不可能。
可以通过包含空单元格的行，或者将列中的所有单元格转换为相同格式的数据。

pd.to_datetime(arg, errors='raise', dayfirst=False, yearfirst=False, utc=False, format=None, unit='ns')
参数说明：
arg：原始数据，需要被转换成日期时间格式。这个参数可以是字符串、列表、Series（一维数据）等。

errors：处理数据中无法解析的日期时的方式：
  'raise'（默认）：遇到问题则报错。
  'coerce'：将无效日期转为 NaT（Not a Time，即非时间）。
  'ignore'：什么都不处理，仅返回原始数据。
  
dayfirst：设置为 True 时，认为日期的格式是“日/月/年”。

yearfirst：设置为 True 时，认为日期格式是“年/月/日”。

utc：设置为 True 时，结果以 UTC 时间返回（需有时间区支持）。

format：用于指定字符串的时间格式，可以避免解析歧义。

unit：当 arg 是数字类型时，指定该数字的单位（如 'D' 表示天数，'s' 表示秒）。
'''
# 示例  格式化日期
import pandas as pd

# 第三个日期格式是错误的
data = {
  "Date": ['2020/12/01', '2020/12/02' , '20201226'],
  "duration": [50, 40, 45]
}

df = pd.DataFrame(data,index = ["day1", "day2", "day3"])

df['Date'] = pd.to_datetime(df['Date'],format = 'mixed')

print(df)
'''
           Date  duration
day1 2020-12-01        50
day2 2020-12-02        40
day3 2020-12-26        45

format='mixed' 作用
当一列里存在多种不统一的日期格式时使用，pandas 自动适配、解析每一条日期，不用手动指定单一format。
如果不写 format='mixed'，直接 pd.to_datetime() 新版本 pandas 也能自动识别，但加上更稳妥。
'''
```

```python
# 清洗错误数据

# 示例   替换错误年龄的数据
import pandas as pd

person = {
  "name": ['Google', 'JD' , 'Taobao'],
  "age": [50, 40, 12345]    # 12345 年龄数据是错误的
}
df = pd.DataFrame(person)

df.loc[2,'age'] = 30

print(df)
'''
     name    age
0  Google     50
1  JD         40
2  Taobao     30

df.loc[行标识, 列名] = 新值
'''
# 设置条件语句     将 age 大于 30 的设置为 30
for x in df.index:
    if df.loc[x,'age'] > 30:
        df.loc[x,'age'] = 30
print(df)
'''
     name  age
0  Google   30
1      JD   30
2  Taobao   30
'''
# 将错误数据的行删除    将 age 大于 120 的删除
person = {
  "name": ['Google', 'JD' , 'Taobao'],
  "age": [50, 40, 12345]    # 12345 年龄数据是错误的
}
df1 = pd.DataFrame(person)
for x in df1.index:
    if df1.loc[x,'age'] > 120:
        df1.drop(x,inplace = True)
print(df1)
'''
     name  age
0  Google   50
1      JD   40
'''
```

```python
# 清洗重复数据
'''
如果要清洗重复数据，可以使用 duplicated() 和 drop_duplicates() 方法。
如果对应的数据是重复的，duplicated() 会返回 True，否则返回 False。
'''
# 示例
import pandas as pd

person = {
  "name": ['Google', 'Runoob', 'Runoob', 'Taobao'],
  "age": [50, 40, 40, 23]  
}
df = pd.DataFrame(person)

print(df.duplicated())
'''
0    False
1    False
2     True
3    False
dtype: bool
'''

# 删除重复数据，可以直接使用drop_duplicates() 方法
df.drop_duplicates(inplace = True)  # inplace = True ，修改并覆盖原数据
print(df)
'''
     name  age
0  Google   50
1  Runoob   40
3  Taobao   23
'''
```

```python
# 独热编码
import pandas as pd

data = {'City': ['New York', 'Los Angeles', 'Chicago', 'Houston']}
df = pd.DataFrame(data)

# 对 "City" 列进行 One-Hot 编码
df_encoded = pd.get_dummies(df,columns = ['City'])

print(df_encoded)
'''
   City_Chicago  City_Houston  City_Los Angeles  City_New York
0         False         False             False           True
1         False         False              True          False
2          True         False             False          False
3         False          True             False          False

False 为 0；True 为 1

pd.get_dummies(data, columns=需要编码的列的列表)
参数解释：
data：DataFrame / Series
columns=[列名1,列名2]：指定只对这几列做编码，其他列原样保留
drop_first=True：去掉第一列，规避多重共线性（机器学习训练必用）
dtype=int：把布尔值 True/False 转为 1/0

字符串分类文本不能直接丢进机器学习模型，独热编码把分类文本转换成 0、1 数值特征，供算法使用。
原本这个数据在那列的第 几 位 那么那个数据独热后的列里第几个数字就为 1 
1 起到「标记、选中」的作用，用来区分样本属于哪一类
'''


# 标准化
from sklearn.preprocessing import StandardScaler
import pandas as pd

data = {'Age': [25, 30, 35, 40, 45],
        'Salary': [50000, 60000, 70000, 80000, 90000]}

df = pd.DataFrame(data)

# 标准化数据
scaler = StandardScaler()
df_scaled = scaler.fit_transform(df)

print(df_scaled)
'''
[[-1.41421356 -1.41421356]
 [-0.70710678 -0.70710678]
 [ 0.          0.        ]
 [ 0.70710678  0.70710678]
 [ 1.41421356  1.41421356]]
 
输出是numpy 二维数组，不是 DataFrame。

标准化公式（核心）
z = (x - u) / p
x ：原始数值
u ：该列平均值 mean
p：该列标准差 std

作用：
每一列处理后：均值 = 0，标准差 = 1
消除量纲影响（比如年龄 25~45，薪资 5 万～9 万，数值差距巨大）
很多机器学习模型（SVM、逻辑回归、神经网络）必须标准化
'''
```
