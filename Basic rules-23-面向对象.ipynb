```python
# 面向对象
'''
Python从设计之初就已经是一门面向对象的语言

面向对象技术简介
类(Class): 用来描述具有相同的属性和方法的对象的集合。它定义了该集合中每个对象所共有的属性和方法。对象是类的实例。
方法：类中定义的函数。
类变量：类变量在整个实例化的对象中是公用的。类变量定义在类中且在函数体之外。类变量通常不作为实例变量使用。
数据成员：类变量或者实例变量用于处理类及其实例对象的相关的数据。
方法重写：如果从父类继承的方法不能满足子类的需求，可以对其进行改写，这个过程叫方法的覆盖（override），也称为方法的重写。
局部变量：定义在方法中的变量，只作用于当前实例的类。
实例变量：在类的声明中，属性是用变量来表示的，这种变量就称为实例变量，实例变量就是一个用 self 修饰的变量。
继承：即一个派生类（derived class）继承基类（base class）的字段和方法。继承也允许把一个派生类的对象作为一个基类对象对待。例如，有这样一个设计：一个Dog类型的对象派生自Animal类，这是模拟"是一个（is-a）"关系（例图，Dog是一个Animal）。
实例化：创建一个类的实例，类的具体对象。
对象：通过类定义的数据结构实例。对象包括两个数据成员（类变量和实例变量）和方法。

Python中的类提供了面向对象编程的所有基本功能：类的继承机制允许多个基类，派生类可以覆盖基类中的任何方法，方法中可以调用基类中的同名方法。
对象可以包含任意数量和类型的数据。
'''
```

```python
# 类定义
'''
语法格式如下：
class ClassName:
    <statement-1>
    .
    .
    .
    <statement-N>
类实例化后，可以使用其属性，实际上，创建一个类之后，可以通过类名访问其属性。

类对象
类对象支持两种操作：属性引用和实例化。
属性引用使用和 Python 中所有的属性引用一样的标准语法：obj.name。
类对象创建后，类命名空间中所有的命名都是有效属性名。
'''
class MyClass:
    """一个简单的类实例"""
    i = 12345
    def f(self):
        return 'hello world'
 
# 实例化类
x = MyClass()
 
# 访问类的属性和方法
print("MyClass 类的属性 i 为：", x.i)
print("MyClass 类的方法 f 输出为：", x.f())
'''
以上创建了一个新的类实例并将该对象赋给局部变量 x，x 为空的对象。
输出结果为：
MyClass 类的属性 i 为： 12345
MyClass 类的方法 f 输出为： hello world
'''

# __init__() 的参数通过 __init__() 传递到类的实例化操作上
class Complex:
    def __init__(self,realpart,imagpart):
        self.realpart = realpart
        self.imagpart = imagpart

x = Complex(3.0,-4.5)
print(x.realpart,x.imagpart)   # 3.0 -4.5
'''
self 代表类的实例，而非类
在 Python中，self 是一个惯用的名称，用于表示类的实例（对象）自身。它是一个指向实例的引用，使得类的方法能够访问和操作实例的属性。
当你定义一个类，并在类中定义方法时，第一个参数通常被命名为 self，尽管你可以使用其他名称，但强烈建议使用 self，以保持代码的一致性和可读性。
'''
```

```python
# 类的方法
'''
在类的内部，使用 def 关键字来定义一个方法
与一般函数定义不同，类方法必须包含参数 self, 且为第一个参数，self 代表的是类的实例。
'''
#类定义
class people:
    #定义基本属性 
    name = ''
    age = 0
    #定义私有属性,私有属性在类外部无法直接进行访问
    __weight = 0
    #定义构造方法
    def __init__(self,n,a,w):
        self.name = n
        self.age = a
        self.__weight = w
    def speak(self):
        print("%s 说: 我 %d 岁。" %(self.name,self.age))
 
# 实例化类
p = people('lion',10,30)
p.speak()
'''
输出：
lion 说: 我 10 岁。

name = ''、age = 0、__weight = 0
写在__init__外面，属于类属性（类变量）
类属性归属类本身，所有实例共享；
后续 self.name = n 会在实例上创建实例属性，遮蔽同名类属性

私有属性 __weight
Python 名称改写机制：以两个下划线开头的属性 __xxx，会被解释器自动改名 _类名__xxx。
表层效果：外部不能直接 实例.__weight 访问，模拟私有；
⚠️ 不是真正安全的封装，只是语法伪装，依然可以 p._people__weight 强行读取。
print(p.__weigh)
直接打印他，会报异常 ：AttributeError: 'people' object has no attribute '__weigh'
'''
```

```python
# 继承
'''
Python 同样支持类的继承，如果一种语言不支持继承，类就没有什么意义。派生类的定义如下所示:
class DerivedClassName(BaseClassName):
    <statement-1>
    .
    .
    .
    <statement-N>
子类（派生类 DerivedClassName）会继承父类（基类 BaseClassName）的属性和方法。
BaseClassName（实例中的基类名）必须与派生类定义在一个作用域内。除了类，还可以用表达式，基类定义在另一个模块中时这一点非常有用:
class DerivedClassName(modname.BaseClassName):
'''

# 类定义
class people:
    # 定义基本属性
    name = ''
    age = 0
    weight = 0

    # 定义构造方法
    def __init__(self,n,a,w):
        self.name = n
        self.age = a
        self.weight = w
    def speak(self):
        print(f"{self.name} 说：我{self.age}岁 ")

# 单继承示例
class student(people):
    grade = ''
    def __init__(self,n,a,w,g):
        # 调用父类的构函
        people.__init__(self,n,a,w)  # 显式调用父类构造函数
        self.grade = g
    # 覆盖父类的方法
    def speak(self):
        print(f"{self.name} 说：我{self.age}岁,正在读 {self.grade}年级")

s = student('ken',10,60,4)
s.speak()
'''
输出：ken 说：我10岁,正在读 4年级
'''
```

```python
# 多继承
'''
简单的总结 多继承就是 一个子类继承了多个父类，属于 一继承多
Python同样有限的支持多继承形式。多继承的类定义形如下例:
class DerivedClassName(Base1, Base2, Base3):
    <statement-1>
    .
    .
    .
    <statement-N>
需要注意圆括号中父类的顺序，若是父类中有相同的方法名，而在子类使用时未指定，python从左至右搜索 即方法在子类中未找到时，从左到右查找父类中是否包含方法。
'''

#类定义
class People:
    def __init__(self, n, a, w):
        self.name = n
        self.age = a
        self.__weight = w

    def speak(self):
        print(f"{self.name} 说: 我 {self.age} 岁。")


# 单继承
class Student(People):
    def __init__(self, n, a, w, g):
        People.__init__(self, n, a, w)
        self.grade = g

    def speak(self):
        print(f"{self.name} 说: 我 {self.age} 岁了，我在读 {self.grade} 年级")


# 另一个类
class Speaker:
    def __init__(self, speaker_name, t):
        self.speaker_name = speaker_name
        self.topic = t

    def speak(self):
        print(f"我叫 {self.speaker_name}，我是一个演说家，我演讲的主题是 {self.topic}")


# 多继承
class Sample(Speaker, Student):
    def __init__(self, name, age, weight, grade, topic):
        Student.__init__(self, name, age, weight, grade)
        Speaker.__init__(self, name, topic)


test = Sample("Tim", 25, 80, 4, "Python")
test.speak()
'''
输出：我叫Tim，我是一个演说家，我今天的演讲主题是Python

逻辑顺序：
1. People 父类
作用：描述普通人，拥有姓名、年龄、私有体重；
self.name：普通人名称；
__weight：私有属性，外部不可直接访问；
speak()：普通人的发言。

2.Student(People) 单继承子类
继承 People，拥有人的全部属性；额外增加年级 grade；
重写 speak()，实现学生专属发言；
self.name 沿用父类定义，代表【学生姓名】。

3.Speaker 独立父类（无继承 People）
用：描述演说家，拥有姓名、演说主题；
self.speaker_name：演说家名称；与之前 People 的 self.name 相区别
speak()：演说家的发言。

4. Sample(Speaker, Student) 多继承核心类
继承顺序：(Speaker, Student)
MRO 查找顺序：Sample → Speaker → Student → People → object
构造执行顺序：
① 执行 Student.__init__
创建：self.name、self.age、self.__weight、self.grade
② 执行 Speaker.__init__
创建：self.speaker_name、self.topic
两个初始化操作读写不同属性，互不干扰，没有覆盖风险。

5. 实例调用
寻找 speak 方法：
Sample 自身无 speak → 查找第一个父类 Speaker，找到 speak，直接执行；
打印演说家信息。

此例的继承关系：
1.Student 的父类：People（单继承）
2.Sample 有两个直接父类：Speaker、Student
3.Student 又间接拥有父类 People
'''
```

```python
# 方法重写
# 如果你的父类方法的功能不能满足你的需求，你可以在子类重写你父类的方法，实例如下：
class Parent:    # 定义父类
    def myMethod(self):
        print ('调用父类方法')
 
class Child(Parent):   # 定义子类
    def myMethod(self):
        print ('调用子类方法')
        super().myMethod()   # 在子类内部调用父类同名方法

c = Child()
c.myMethod()
'''
输出：
调用子类方法
调用父类方法

super() 函数是用于调用父类(超类)的一个方法
super () 是拿到【MRO 序列中，当前类的下一个类】
单继承场景下，“下一个类” 刚好就是父类，所以是找父类；

两种 super 语法
1）类内部简写
super().myMethod() 等价自动填充：
super(Child, self).myMethod()
第一个参数：当前所在类（Child）
第二个参数：实例对象（self）
类内部，可以简写 super()

2）完整显式写法（类内、类外都能用)
super(Child, c).myMethod()
类外部，不能简写！必须传两个参数

uper (A, obj) 两条核心规则
Python 先获取 obj 的 MRO 列表（方法查找顺序链）
在 MRO 列表里，找到类 A，取 A 后面紧邻的那一个类，这就是 super 拿到的对象。
'''

# super () 在 __init__ 构造方法中的标准用法
'''
最重要的一条规则：
如果子类重写了 __init__，不会自动调用父类的 __init__，必须手动触发；
优先使用 super()，不要硬编码 父类.__init__(self,...)。
'''
# 单继承标准示范

class Parent:
    def __init__(self, name):
        self.name = name
        print("父类构造执行")

class Child(Parent):
    def __init__(self, name, age):
        # super简写形式，等价 super(Child, self).__init__(name)
        super().__init__(name)
        self.age = age
        print("子类构造执行")

c = Child("小明", 12)
print(c.name, c.age)
'''
输出：
父类构造执行
子类构造执行
小明 12

super().__init__(name)
super() 自动识别当前类是 Child，实例是 self
沿着 MRO 找到 Child 下一个类：Parent
执行 Parent.__init__(self, name)
不要传 self！super 内部自动把 self 传入父类构造
'''

# 多继承案例

class People:
    def __init__(self, name):
        self.name = name

class Student(People):
    def __init__(self, name, grade):
        super().__init__(name)
        self.grade = grade

class Speaker:
    def __init__(self, topic):
        self.topic = topic

# 多继承
class Sample(Student, Speaker):
    def __init__(self,name,grade,topic):
        super().__init__(name, grade)
        # super链条走到Student结束，不会自动走到Speaker
        # 这种并行父类，需要额外手动初始化
        Speaker.__init__(self, topic)

s = Sample("Tim",4,"Python")
print(s.name, s.grade, s.topic)
'''
输出 : Tim 4 Python
'''
'''
MRO 链表顺序优先级： 子类自身定义 > MRO 顺序靠前父类 > MRO 靠后父类 > object
'''
```
