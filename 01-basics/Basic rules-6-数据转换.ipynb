```python
#数据转换
"""
Python 数据类型转换可以分为两种：
#隐式类型转换 - 自动完成
#显式类型转换 - 需要使用类型函数来转换

#隐式类型转换
#在隐式类型转换中，Python 会自动将一种数据类型转换为另一种数据类型，不需要我们去干预。
"""
num_int = 123   #int 整型
num_flo = 1.23  #float 浮点型

num_new = num_int + num_flo   #整型 + 浮点型 为浮点型，这是因为 Python 会将较小的数据类型转换为较大的数据类型，以避免数据丢失。

print("num_int 数据类型为:",type(num_int))
print("num_flo 数据类型为:",type(num_flo))
print("num_new 值为:",num_new)
print("num_new 数据类型为:",type(num_new))

num_int = 123
num_str = "456"

print("num_int 数据类型为:",type(num_int))
print("num_str 数据类型为:",type(num_str))

#print(num_int+num_str)  这种输出会报错：TypeError: unsupported operand type(s) for +: 'int' and 'str'
#因为整型和字符串类型就无法隐式转换，得用显式
```

```python
#显示转换
#在显式类型转换中，用户将对象的数据类型转换为所需的数据类型。 
#我们使用 int()、float()、str() 等预定义函数来执行显式类型转换。

#int() 强制转换为整型：
x = int(1)   # x 输出结果为 1
y = int(2.8) # y 输出结果为 2
z = int("3") # z 输出结果为 3
print("x的值为：",x,"x的类型为：",type(x))
print("y的值为：",y,"y的类型为：",type(y))
print("z的值为：",z,"z的类型为：",type(z))

#float() 强制转换为浮点型：
a = float(1)     # x 输出结果为 1.0
b = float(2.8)   # y 输出结果为 2.8
c = float("3")   # z 输出结果为 3.0
d = float("4.2") # w 输出结果为 4.2
print("a的值为：",a,"a的类型为：",type(a))
print("b的值为：",b,"b的类型为：",type(b))
print("c的值为：",c,"c的类型为：",type(c))
print("d的值为：",d,"d的类型为：",type(d))

#str() 强制转换为字符串类型：
e = str("s1") # x 输出结果为 's1'
f = str(2)    # y 输出结果为 '2'
g = str(3.0)  # z 输出结果为 '3.0'
print("e的结果为：",e,"e的类型为：",type(e))
print("f的结果为：",f,"f的类型为：",type(f))
print("g的结果为：",g,"g的类型为：",type(g))
```

```python
#整型和字符串类型进行运算
num_int = 123
num_str= "456"                    #因为字符串里的内容就是数字，所以可以进行转换
print("num_int 数据类型为：",type(num_int))
print("类型转换前，num_str 数据类型为：",type(num_str))

num_str = int(num_str)      #强制类型转换
print("类型转换后，num_str 数据类型为：",type(num_str))

num_sum = num_int + num_str
print("num_int 与 ：num_str 相加的结果",num_sum)
print("num_sum 数据类型为：",type(num_sum))

#你可以轻松地将整数转换为字符串，因为每一个整数都有一个明确的字符串表示（例如，整数123可以表示为字符串"123"）。
#对于一个非数字字符串（如"Hello"），它无法被转换为一个整数或浮点数，因为这个字符串并不包含任何可以表示一个数字的信息。
#对于一个列表或元组，它可以被转换为一个集合（如果它的元素是不可变的），但不能被转换为一个整数，因为一个集合或列表中的元素无法合理地表示为一个单独的数字。
```
