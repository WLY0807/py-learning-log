```python
import sys; x = 'runoob'; num = sys.stdout.write(x + '\n')    # import sys：导入 Python 内置标准库sys，提供系统相关操作。
print("字节总数:",num)                                         #sys.stdout：标准输出流对象，print()底层本质就是封装了sys.stdout.write。
                                                        #sys.stdout.write(字符串)：直接向控制台写入文本，不会自动换行，所以手动拼接 \n 实现换行。
                                                        #Python 可以在同一行中使用多条语句，语句之间使用分号 ; 分割。
                                                        #输出结果为 7 ，此处的 7 表示字符数，runoob 有 6 个字符，\n 表示一个字符，加起来 7 个字符。
input("\n\n按下enter键后退出")    # \n\n 在结果输出前会输出两个新的空行。一旦用户按下 enter 键时，程序将退出。
                                 #input() 是 Python 内置函数，作用：暂停程序，接收用户从键盘输入的一行文本，按下回车后把输入内容以字符串形式返回。
```
