```python
# 正则表达式
'''
什么是正则表达式
正则表达式（Regular Expression，常简写为 regex、re）是一套跨编程语言的通用文本匹配规则语言。
它的本质是：你用一组特殊符号写好「筛选规则」，程序就可以在一大段文本中，自动完成「查找、提取、校验、替换」所有符合规则的内容。

正则能解决什么问题
日常开发中，所有和「文本处理」相关的需求，正则都能大幅提升效率，核心场景 4 类：
格式校验：判断用户输入的手机号、邮箱、密码是否符合规则
内容提取：从一大段杂乱文本中，批量提取所有手机号、链接、日期、关键数据
批量替换：统一替换文本中符合规则的内容，比如脱敏手机号、统一日期格式
文本清洗：去除文本中的多余空格、特殊符号、无效换行

前置约定：原始字符串 r""
正则规则在 Python 中，统一推荐加r前缀，写成r"规则"，和之前的文件路径原始字符串用法完全一致。
作用：避免 Python 字符串转义和正则转义产生冲突，比如正则里的\d，不加 r 可能被 Python 当成转义字符处理
'''

```

```python
# 正则表达式模式
'''
式字符串使用特殊的语法来表示一个正则表达式。
字母和数字表示他们自身。一个正则表达式模式中的字母和数字匹配同样的字符串。
多数字母和数字前加一个反斜杠时会拥有不同的含义。
标点符号只有被转义时才匹配自身，否则它们表示特殊的含义。
反斜杠本身需要使用反斜杠转义。
由于正则表达式通常都包含反斜杠，所以最好使用原始字符串来表示它们。

下表列出了正则表达式模式语法中的特殊元素。如果你使用模式的同时提供了可选的标志参数，某些模式元素的含义会改变。
模式	        描述
^	            匹配字符串的开头
$	            匹配字符串的末尾。
.	            匹配任意字符，除了换行符，当re.DOTALL标记被指定时，则可以匹配包括换行符的任意字符。
[...]	        用来匹配所包含的任意一个字符，例如 [amk] 匹配 'a'，'m'或'k'
[^...]	        不在[]中的字符：[^abc] 匹配除了a,b,c之外的字符。
re*	            匹配0个或多个的表达式。
re+	            匹配1个或多个的表达式。
re?	            匹配0个或1个由前面的正则表达式定义的片段，非贪婪方式
re{ n}	        匹配n个前面表达式。例如，"o{2}"不能匹配"Bob"中的"o"，但是能匹配"food"中的两个o。
re{ n,}	        精确匹配n个前面表达式。例如，"o{2,}"不能匹配"Bob"中的"o"，但能匹配"foooood"中的所有o。"o{1,}"等价于"o+"。"o{0,}"则等价于"o*"。
re{ n, m}	    匹配 n 到 m 次由前面的正则表达式定义的片段，贪婪方式
a| b	        匹配a或b
(re)	        匹配括号内的表达式，也表示一个组
(?imx)	        正则表达式包含三种可选标志：i, m, 或 x 。只影响括号中的区域。
(?-imx)	        正则表达式关闭 i, m, 或 x 可选标志。只影响括号中的区域。
(?: re)	        类似 (...), 但是不表示一个组
(?imx: re)	    在括号中使用i, m, 或 x 可选标志
(?-imx: re)	    在括号中不使用i, m, 或 x 可选标志
(?#...)	        注释.
(?= re)	        前向肯定界定符。如果所含正则表达式，以 ... 表示，在当前位置成功匹配时成功，否则失败。但一旦所含表达式已经尝试，匹配引擎根本没有提高；模式的剩余部分还要尝试界定符的右边。
(?! re)	        前向否定界定符。与肯定界定符相反；当所含表达式不能在字符串当前位置匹配时成功。
(?> re)	        匹配的独立模式，省去回溯。
\w	            匹配数字字母下划线
\W	            匹配非数字字母下划线
\s	            匹配任意空白字符，等价于 [\t\n\r\f]。
\S	            匹配任意非空字符
\d	            匹配任意数字，等价于 [0-9]。
\D	            匹配任意非数字
\A	            匹配字符串开始
\Z	            匹配字符串结束，如果是存在换行，只匹配到换行前的结束字符串。
\z	            匹配字符串结束
\G	            匹配最后匹配完成的位置。
\b	            匹配一个单词边界，也就是指单词和空格间的位置。例如， 'er\b' 可以匹配"never" 中的 'er'，但不能匹配 "verb" 中的 'er'。
\B	            匹配非单词边界。'er\B' 能匹配 "verb" 中的 'er'，但不能匹配 "never" 中的 'er'。
\n, \t, 等。	匹配一个换行符。匹配一个制表符, 等
\1...\9	        匹配第n个分组的内容。
\10	            匹配第n个分组的内容，如果它经匹配。否则指的是八进制字符码的表达式。
'''
```

```python
# 正则表达式 - 语法
'''
正则的所有规则，本质上都是在回答三个问题：匹配什么字符、匹配多少个、匹配在什么位置。

1.第一层：单字符匹配（匹配什么字符）
单字符规则是正则的最小单位，一条单字符规则，对应匹配文本里的 1 个字符。
1.1 普通字符：字面完全匹配
没有特殊含义的字母、数字、汉字，都是普通字符，规则写什么，就精准匹配什么内容。
'''
import re
text = 'python学习    python入门  python'
rule = r'python'
print(re.findall(rule, text))
# 输出：['python', 'python', 'python']   ;精准匹配所有"python"字符串

'''
1.2 元字符：特殊含义的单字符匹配
元字符是正则里有特殊功能的符号，不用记全，先掌握最常用的 6 个：
元字符	匹配规则	                               示例规则	   能匹配的内容
.	    匹配任意 1 个字符（除换行符\n外）	       r"a.b"	   a1b、axb、a@b、a b
\d      匹配1 个数字（0-9）	                   r"\d\d"	   12、34、99（两位数字）
\w      匹配1 个字母 / 数字 / 下划线	           r"\w\w"	   ab、12、a_、测试（中文也匹配）
\s      匹配1 个空白字符（空格、制表符、换行符）   r"a\sb"	   a b（中间有空格）、a 换行 b

1.3 反义元字符（大写版本，和小写反过来）
元字符	匹配规则
\D      匹配 1 个非数字的字符
\W      匹配 1 个非字母 / 数字 / 下划线的字符
\S      匹配 1 个非空白的字符
'''
text_1 = "账号：abc123，密码：456def"
rule = r"\D+"
print(re.findall(rule, text_1))
# 输出：['账号：abc', '，密码：', 'def']

'''
1.4 自定义字符集 []
用方括号自定义匹配范围，匹配方括号内的任意 1 个字符，灵活度最高。
[abc]：匹配 a、b、c 中的任意一个
[a-z]：匹配任意一个小写字母
[A-Z]：匹配任意一个大写字母
[0-9]：匹配任意一个数字，和\d效果完全一致
[a-zA-Z0-9]：匹配任意一个字母或数字
[^abc]：方括号开头加^，代表取反，匹配除了 a、b、c 之外的任意字符
'''
# 示例：匹配第二位是 3/5/8 的手机号开头
text_2 = "138xxx 159xxx 123xxx 176xxx"
rule = r"1[358]\d"
print(re.findall(rule, text_2))
# 输出：['138', '159']，只有第二位在3/5/8范围内的才会被匹配

'''
2 第二层：量词（匹配多少个字符
上面的单字符规则，默认只能匹配 1 个字符。量词用来控制「前面的字符 / 规则，连续出现多少次」，是正则高频必学内容。
2.1 基础量词
量词	匹配规则	                            示例规则	    匹配说明
+       前面的规则出现1 次及以上（至少 1 次）	r"\d+"	    匹配连续数字：3、12、12345
*       前面的规则出现0 次及以上（可有可无）	r"ab*c"	    匹配 ac、abc、abbc、abbbc
?       前面的规则出现0 次或 1 次（最多 1 次）	r"https?"	匹配 http、https
{n}	    前面的规则精确出现 n 次	            r"\d{11}"	匹配刚好 11 位的数字
{n,m}	前面的规则出现n 到 m 次	            r"\d{3,5}"	匹配 3-5 位连续数字
{n,}	前面的规则出现至少 n 次	            r"\d{6,}"	匹配 6 位及以上的连续数字
最常用的是+、?、{n}三个
'''
# 示例：匹配国内手机号（1 开头、第二位 3-9、总共 11 位数字）
text_3 = "我的手机号13812345678，备用号15987654321，短号12345"
rule = r"1[3-9]\d{9}"
print(re.findall(rule, text_3))
# 输出：['13812345678', '15987654321']，精准匹配11位标准手机号

'''
2.2 贪婪匹配与非贪婪匹配
正则默认是贪婪匹配：量词会尽可能多地匹配内容，能匹配多长就匹配多长。
在量词后面加?，就变成非贪婪匹配：尽可能少地匹配内容，符合规则就停止。
最典型的对比：.* 和 .*?
.匹配任意字符，*匹配 0 次及以上，.*就是贪婪匹配，能吞多长吞多长
.*?是非贪婪匹配，遇到符合规则的结尾就立刻停止
'''
import re
text_4 = '姓名:"张三" 年龄:"18" 城市:"北京"'

# 贪婪匹配：.* 会从第一个引号，一直匹配到最后一个引号
greedy_rule = r'".*"'
print(re.findall(greedy_rule, text_4))
# 输出：['"张三" 年龄:"18" 城市:"北京"']，一整段全匹配了

# 非贪婪匹配：.*? 遇到下一个引号就停止，分段匹配
lazy_rule = r'".*?"'
print(re.findall(lazy_rule, text_4))
# 输出：['"张三"', '"18"', '"北京"']，精准匹配每一段
# 需要分段提取内容时，优先用非贪婪匹配，避免匹配内容过长。

'''
3 第三层：位置匹配（匹配在什么位置）
位置规则不匹配具体的字符，只匹配「字符所在的位置」，用来限定规则必须出现在开头、结尾、单词边界等位置。
3.1 开头与结尾
符号	匹配规则
^       匹配整个字符串的开头
$       匹配整个字符串的结尾
最核心的用途：格式校验，确保整个字符串完全符合规则，没有多余内容。
'''
# 示例：校验输入的是不是纯 6 位数字
# 正确的纯6位数字
text1 = "123456"
# 带多余字符的内容
text2 = "a123456b"

# 制定规则
rule = r"^\d{6}$"

print(re.search(rule, text1))  # <re.Match object; span=(0, 6), match='123456'>  ;有匹配结果，符合规则
print(re.search(rule, text2))  # None，不符合规则
# 如果不加 ^ $ ，\d{6}只要文本里有 6 位连续数字就会匹配，无法实现严格校验。

'''
3.2 单词边界 \b
\b 匹配单词的边界（单词和空格、标点、开头结尾的交界处），用来精准匹配独立单词，避免部分命中。
示例：匹配独立的单词py，不匹配python里的 py
'''
# 示例：匹配独立的单词py，不匹配 python里的 py
text_4 = "py 学习 python 入门 py"
rule = r"\bpy\b"

print(re.findall(rule, text_4))
# 输出：['py', 'py']，只匹配独立的py，不会命中python里的py
# 对应的反义是\B，代表非单词边界。

'''
4 第四层：分支与分组（复杂规则组合）
4.1 分支 |
代表「或者」的关系，匹配竖线左边的规则，或者右边的规则。
'''
text_5 = "aaa@qq.com bbb@163.cn ccc@gmail.org"
rule = r"\w+@\w+\.(com|cn)"
print(re.findall(rule, text_5))
# 输出：['com', 'cn']，这里涉及分组特性

'''
4.2 分组 ()
用圆括号把一段规则打包成一个整体，有两个核心作用：
把多个规则当成一个整体，配合量词使用
捕获分组内容：单独提取括号里的内容，是数据提取的核心功能
作用 1：整体控制量词
'''
# 示例：匹配连续出现 3 次的ab
text_6 = "ababab abc abab"
# 不加括号：b重复3次，匹配abbb
# 加括号：ab整体重复3次
rule = r"(ab){3}"
print(re.search(rule, text_6))
# 匹配到ababab

'''
作用 2：捕获分组（提取核心内容）
正则里如果加了括号，findall函数就只会返回括号里捕获的内容，这是提取数据最常用的技巧。
'''
text_7 = "zhangsan@qq.com lisi@163.cn"
# 括号只包裹用户名部分，只提取这一段
rule = r"(\w+)@\w+\.\w+"
print(re.findall(rule, text_7))
# 输出：['zhangsan', 'lisi']，只返回括号里捕获的内容

# 多个括号就会按顺序返回多个分组内容，示例：提取邮箱的用户名和域名
rule = r"(\w+)@(\w+\.\w+)"
print(re.findall(rule, text_7))
# 输出：[('zhangsan', 'qq.com'), ('lisi', '163.cn')]

'''
补充：非捕获组 (?:)
如果括号只是为了打包整体，不需要提取内容，就写成(?:规则)，不会被捕获返回。
'''
# 示例：匹配.com 或.cn 结尾的完整邮箱，不需要单独返回后缀
rule = r"\w+@\w+\.(?:com|cn)"
print(re.findall(rule, text_7))
# 输出：['zhangsan@qq.com', 'lisi@163.cn']，返回完整邮箱，不是只返回后缀

'''
5 第五层：转义符 \
如果要匹配正则里的元字符本身（比如.、*、+、?、()），必须在前面加\转义，告诉程序「这个符号不是元字符，就是匹配它本身」。
'''
text_8 = "价格12.5元 数量3个"
# 不加转义：.匹配任意字符，会匹配12.5里的2.5
# 加转义：\.只匹配小数点本身
rule = r"\d+\.\d+"
print(re.findall(rule, text_8))
# 输出：['12.5']
```

```python
# Python re 模块
# Python 通过re模块调用正则
'''
re.match函数
re.match 尝试从字符串的起始位置匹配一个模式，如果不是起始位置匹配成功的话，match() 就返回 None。
函数语法：
re.match(pattern, string, flags=0)
函数参数说明：
参数	描述
pattern	匹配的正则表达式
string	要匹配的字符串。
flags	标志位，用于控制正则表达式的匹配方式，如：是否区分大小写，多行匹配等等。参见：正则表达式修饰符 - 可选标志
匹配成功 re.match 方法返回一个匹配的对象，否则返回 None。

我们可以使用 group(num) 或 groups() 匹配对象函数来获取匹配表达式。
匹配对象方法	    描述
group(num=0)	匹配的整个表达式的字符串，group() 可以一次输入多个组号，在这种情况下它将返回一个包含那些组所对应值的元组。
groups()	    返回一个包含所有小组字符串的元组，从 1 到 所含的小组号。
'''
# 实例1
import re
print(re.match('www','www.runoob.com').span())
print(re.match('com','www.runoob.com'))
'''
.span() 方法
span() 是 Match 对象的内置方法，返回匹配到的内容在原文本中的起始索引、结束索引，格式是元组(起始位置, 结束位置)。
Python 索引从 0 开始计数，遵循「左闭右开」规则：包含起始位置的字符，不包含结束位置的字符
匹配到的www是文本的前 3 个字符，对应索引0、1、2，结束索引是3
因此最终输出 (0, 3)
补充：如果想获取匹配到的内容本身，用.group()方法
'''

# 实例2
line = "Cats are smarter than dogs"

matchobj = re.match( r'(.*) are (.*?) .*',line,re.M|re.I)
'''
用 re.match 从字符串开头匹配整行文本，通过两个圆括号分组，
分别提取 are 前面的主语、are 后面的第一个形容词

正则片段	   规则含义	                                                          在本句中匹配的内容
(.*)	   第一个分组（捕获组），.匹配任意非换行字符，*代表 0 次及以上，贪婪匹配	  匹配到第一个 are 之前的所有内容：Cats
are	       普通字符（前后带空格），精准匹配字面内容	                              精准匹配字符串里的空格 + are + 空格
(.*?)	   第二个分组（捕获组），.*?是非贪婪匹配，尽可能短地匹配	                  匹配到后面第一个空格就停止，结果是：smarter
           普通空格字符，精准匹配	                                              匹配 smarter 后面的空格
.*	       贪婪匹配剩余所有非换行字符	                                          匹配剩下的全部内容：than dogs

第一个分组 (.*) 是贪婪匹配，但后面跟着固定的 are，所以它会自动停在第一个 are 前面，最终只匹配到 Cats。
第二个分组 (.*?) 是非贪婪匹配，后面跟着一个空格，所以它遇到第一个空格就立刻停止，只匹配到 smarter

修饰符 re.M|re.I
两个修饰符用竖线拼接，同时生效：
re.I（忽略大小写）：匹配时不区分英文字母大小写，本句中Cats首字母大写，规则里是小写字符也能匹配，是这里唯一生效的修饰符。
re.M（多行模式）：只对^和$起作用，改变首尾匹配的范围；本句正则没有用^/$，且只有一行文本，因此这个参数在这里没有实际效果，属于示例的冗余写法。
'''
if matchobj:
    print("matchobj.group() :",matchobj.group())
    print("matchobj.group(1) :",matchobj.group(1))
    print("matchobj.group(2) :",matchobj.group(2))
else:
    print("No match!")
'''
输出：
matchobj.group() : Cats are smarter than dogs
matchobj.group(1) : Cats
matchobj.group(2) : smarter
'''
```

```python
# re.search()
'''
re.search 扫描整个字符串并返回第一个成功的匹配。
函数语法：
re.search(pattern, string, flags=0)

函数参数说明：
参数	描述
pattern	匹配的正则表达式
string	要匹配的字符串。
flags	标志位，用于控制正则表达式的匹配方式，如：是否区分大小写，多行匹配等等。参见：正则表达式修饰符 - 可选标志
匹配成功re.search方法返回一个匹配的对象，否则返回None。
我们可以使用group(num) 或 groups() 匹配对象函数来获取匹配表达式。

匹配对象方法	    描述
group(num=0)	匹配的整个表达式的字符串，group() 可以一次输入多个组号，在这种情况下它将返回一个包含那些组所对应值的元组。
groups()	    返回一个包含所有小组字符串的元组，从 1 到 所含的小组号。
'''
# 实例
import re 
print(re.search('www','www.google.com').span(), end = '  ')  # 在起始位置匹配
print(re.search('com','www.google.com').span())  # 不在起始位置匹配
# 输出：(0, 3)  (11, 14)
# span() 是 Match 对象的内置方法，返回匹配到的内容在原文本中的起始索引、结束索引，格式是元组(起始位置, 结束位置)。
# Python 索引从 0 开始计数，遵循「左闭右开」规则：包含起始位置的字符，不包含结束位置的字符


# 实例2
line = "Kobe is stronger than James"

searchObj = re.search(r'(.*) is (.*?) .*',line,re.M|re.I)
if searchObj:
    print("searchObj.group():",searchObj.group())
    print("searchObj.group(1):",searchObj.group(1))
    print("searchObj.group(2):",searchObj.group(2))
else:
    print("No match!!")
'''
输出：
searchObj.group(): Kobe is stronger than James
searchObj.group(1): Kobe
searchObj.group(2): stronger
'''

'''
re.match 与 re.search的区别
re.match 只匹配字符串的开始，如果字符串开始不符合正则表达式，则匹配失败，函数返回 None，而 re.search 匹配整个字符串，直到找到一个匹配。
'''
line_1 = "Cats are smarter than dogs"

match = re.match(r'dogs',line_1,re.M|re.I)
if match:
    print("match ---> ",match.group())
else:
    print("No match")

match_1 = re.match(r'Cats',line_1,re.M|re.I)
if match_1:
    print("match ---> ",match_1.group())
else:
    print("No match")

match_2 = re.search(r'dogs',line_1,re.M|re.I)
if match_2:
    print("search ---> ",match_2.group())
else:
    print("No search!!")
'''
输出：
No match
match --->  Cats
search --->  dogs
'''
```

```python
# 检索和替换
'''
Python 的re模块提供了re.sub用于替换字符串中的匹配项。
语法：
re.sub(pattern, repl, string, count=0, flags=0)

参数：
pattern : 正则中的模式字符串。
repl : 替换的字符串，也可为一个函数。
string : 要被查找替换的原始字符串。
count : 模式匹配后替换的最大次数，默认 0 表示替换所有的匹配。
flags : 编译时用的匹配模式，数字形式。
前三个为必选参数，后两个为可选参数。
'''

# 实例
import re

phone = "2026-07-20  # 这是一串电话号码"

# 删除注释部分
num = re.sub(r'#.*$','',phone)
print("电话号码：",num)

# 删除非数字的内容
num_1 = re.sub(r'\D','',phone)
print("电话号码：",num_1)
'''
输出：
电话号码： 2026-07-20  
电话号码： 20260720
'''

# 当 repl 是函数时
# 以下实例中将字符串中的匹配的数字乘以 2
def func_2(match):
    value = int(match.group('func'))
    return str(value*2)

s = '678BHFSIV213'
print(re.sub('(?P<func>\d+)',func_2,s))
# 1356BHFSIV426
'''
re.sub('(?P<value>\d+)'
正则里 () 叫捕获分组，作用：把括号里匹配到的内容单独存下来，方便后续提取。
( 后面紧跟 ?，代表「这个分组不是普通捕获分组，有特殊功能」

只要括号一开，紧接着写 ?，就代表修改分组默认行为，不是命名分组专属，有好多种：
(?:xxx) —— ?: 非捕获分组，只匹配不保存分组，不能 group 取值
(?=xxx) —— ?= 正向预查，只判断不吞字符
(?P<name>xxx) —— ?P Python 专属命名捕获分组
这里的 ? 统一作用：标识当前括号是特殊功能分组。
没有 ? 的 (xxx) = 普通数字编号捕获分组；
有 ? 的 (?xxx) = 特殊功能分组。

只有想给分组自定义名字时，才写 ?P<xxx>，P 只是命名分组的专属标识，别的分组不需要。

(?P<name>) = 给分组起名（定义）
match.group('name') = 通过名字拿内容（使用）
定义必须存在，使用才能生效；
所以这里的 .group('func') 能不报错是因为 (?P<func>\d+) 这里给了 func 的定义，哪怕是前者先出现。代码执行顺序和书写顺序不一定一样
代码上下书写顺序不改变「先定义、后使用」的底层逻辑。

这段代码不用这个命名，用原始默认的数字编号也是能得出同样结果，起一个示范作用
'''

def func_2(match):
    value = int(match.group(1))  # 因为规则里只有一个括号 (\d+) ，所以.group()里为 1 
    return str(value*2)

s = '678BHFSIV213'
print(re.sub('(\d+)',func_2,s))
# 1356BHFSIV426
```

```python
# re.compile() 函数
'''
.compile() 函数用于编译正则表达式，生成一个正则表达式（ Pattern ）对象，供 match() 和 search() 这两个函数使用。
语法格式为：
re.compile(pattern[, flags])
参数：
pattern : 一个字符串形式的正则表达式
flags 可选，表示匹配模式，比如忽略大小写，多行模式等，具体参数为：

模式：
re.IGNORECASE 或 re.I - 使匹配对大小写不敏感
re.L 表示特殊字符集 \w, \W, \b, \B, \s, \S 依赖于当前环境
re.MULTILINE 或 re.M - 多行模式，改变 ^ 和 $ 的行为，使它们匹配字符串的每一行的开头和结尾。
re.DOTALL 或 re.S - 使 . 匹配包括换行符在内的任意字符。
re.ASCII - 使 \w, \W, \b, \B, \d, \D, \s, \S 仅匹配 ASCII 字符。
re.VERBOSE 或 re.X - 忽略空格和注释，可以更清晰地组织复杂的正则表达式。
这些标志可以单独使用，也可以通过按位或（|）组合使用。例如，re.IGNORECASE | re.MULTILINE 表示同时启用忽略大小写和多行模式。
'''
import re
pattern = re.compile(r'\d+')   # 用于匹配至少一个数字
m = pattern.match('one12twothree34four')   # 查找头部，没有匹配
print(m)   # None
'''
pattern.match(string, pos=0, endpos=len(string))
match() 规则：只从指定起始位置向后，开头必须立刻匹配上，中间字符不会跳过
pos：检索起始下标；endpos：检索截止下标（取不到）
不传 pos，默认从下标 0 开始匹配，第一个字符是字母o，不是数字
match() 要求当前起点必须直接匹配规则，开头不满足直接返回 None
'''

m = pattern.match('one12twothree34four',2,10)  # 从'e'的位置开始匹配，没有匹配
print(m)   # None
'''
pos=2，从下标 2 字符e开始匹配
起点是字母，不是数字，匹配失败 → None
'''

m = pattern.match('one12twothree34four', 3, 10)  # 从'1'的位置开始匹配，正好匹配
print(m)  # <re.Match object; span=(3, 5), match='12'>
'''
pos=3，起点字符是1，符合\d+数字规则
连续数字是1和2，对应下标 3、4；结束下标为 5（endpos 取不到）
匹配成功，返回 Match 对象，span 显示匹配区间 (3,5)，匹配内容12
'''

m.group(0)   # '12' ;group(0)：固定获取本次完整匹配到的字符串

m.start(0)   # 3  ;返回匹配内容起始下标，这里数字 1 在索引 3 → 输出3

m.end(0)     # 5  ;返回匹配内容结束下标（不包含），数字 2 在索引 4，结束位置是 5 → 输出5

m.span(0)    # (3, 5)  ;等价于 (start(0), end(0))，一次性返回起止下标元组 → (3, 5)


'''
在上面，当匹配成功时返回一个 Match 对象，其中：
group([group1, …]) 方法用于获得一个或多个分组匹配的字符串，当要获得整个匹配的子串时，可直接使用 group() 或 group(0)；
start([group]) 方法用于获取分组匹配的子串在整个字符串中的起始位置（子串第一个字符的索引），参数默认值为 0；
end([group]) 方法用于获取分组匹配的子串在整个字符串中的结束位置（子串最后一个字符的索引+1），参数默认值为 0；
span([group]) 方法返回 (start(group), end(group))。

'''
```

```python
#re.findall()
'''
在字符串中找到正则表达式所匹配的所有子串，并返回一个列表，如果有多个匹配模式，则返回元组列表，如果没有找到匹配的，则返回空列表。
注意： match 和 search 是匹配一次 findall 匹配所有。
语法格式为：
re.findall(pattern, string, flags=0)
或
pattern.findall(string[, pos[, endpos]])

参数：
pattern   匹配模式。
string    待匹配的字符串。
pos       可选参数，指定字符串的起始位置，默认为 0。
endpos    可选参数，指定字符串的结束位置，默认为字符串的长度。
'''
# 实例
# 查找字符串中的所有数字：
import re

result1 = re.findall(r'\d+','google 456 edge 789')

pattern = re.compile(r'\d+')  # 查找数字
result2 = pattern.findall('google 456 edge 789')
result3 = pattern.findall('run888google123',0,10)  # 在 0：字母 r 与 10： l 之间进行筛选

print(result1)   # ['456', '789']
print(result2)   # ['456', '789']
print(result3)   # ['888']

# 多个匹配模式，返回元组列表：
result = re.findall(r'(\w+)=(\d+)','set width=20 and height=10')
print(result)  # [('width', '20'), ('height', '10')]
```

```python
# re.finditer()
'''
和 findall 类似，在字符串中找到正则表达式所匹配的所有子串，并把它们作为一个迭代器返回。
re.finditer(pattern, string, flags=0)
参数	描述
pattern	匹配的正则表达式
string	要匹配的字符串。
flags	标志位，用于控制正则表达式的匹配方式，如：是否区分大小写，多行匹配等等。
'''

# 实例
import re

it = re.finditer(r"\d+","12a32bc43jf3")
for match in it:
    print(match.group(),end = '   ')
# 12   32   43   3  
```

```python
# re.split()
'''
split 方法按照能够匹配的子串将字符串分割后返回列表，它的使用形式如下：
re.split(pattern, string[, maxsplit=0, flags=0])

参数	     描述
pattern	     匹配的正则表达式
string	     要匹配的字符串。
maxsplit	 分割次数，maxsplit=1 分割一次，默认为 0，不限制次数。
flags	     标志位，用于控制正则表达式的匹配方式，如：是否区分大小写，多行匹配等等。
'''

# 实例
import re 
re.split('\W+','google google google.')   # ['google', 'google', 'google', '']

re.split('(\W+)','google google google.')  # ['google', ' ', 'google', ' ', 'google', '.', '']

re.split('\W+',' google, google, google.',1)  # ['', 'google, google, google.']

re.split(r'a*', 'hello world')  # ['', 'h', 'e', 'l', 'l', 'o', ' ', 'w', 'o', 'r', 'l', 'd', '']
```
