```python
def is_valid_identifier(name):  #创建一个接收字符串参数 name、返回布尔值 True/False 的判断函数。
    try:                        #程序会执行 try 缩进里的所有代码；如果里面代码运行报错，不会直接崩溃，而是跳到下方 except 代码块执行。
        exec(f"{name} = None")  
                                #第一层：f-string 作用：把变量 name 的内容塞进字符串里；等同于如下代码形式：
                                #name = "var2"
                                #s = f"{name} = None"#  None 是 Python 空值。变量名 = None 就是最简单、不会产生副作用的赋值语句，我们只是用来测试语法合不合法，变量存什么值无所谓。
                                #print(s)
                                #第二层：exec() 是 Python 内置函数，功能：接收一段字符串，把字符串里的内容当作代码直接运行；等同于如下代码形式：
                                #exec("a = s") 执行a = s
                                #print(a)  
        return True
    except:                     #和前面 try 配对：只要 try 内部任意代码抛出异常（语法错误、运行错误等），立刻跳过 try 剩余代码，执行 except 里的内容。
        return False

print(is_valid_identifier("2var"))  # False，因为是以数字为开头的字符串，不合法
print(is_valid_identifier("var2"))  # True，以字母和 _ 为开头字符串的是合法的
```
